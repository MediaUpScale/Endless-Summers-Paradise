### Initial Drive Setup & Factory Base Path

````python
import os
from google.colab import drive

def setup_drive_and_factory():
    """
    Monta o Drive e localiza a pasta MEDIAUPSCALE_FACTORY dentro de Colab Notebooks.
    Caminho validado via print do utilizador.
    """
    try:
        # 1. Montagem do Drive
        if not os.path.exists('/content/drive'):
            print("🚀 Montando Google Drive...")
            drive.mount('/content/drive')
        else:
            print("✅ Google Drive já está montado.")

        # 2. Caminho exacto baseado no print image_66b2c2.png
        # O atalho está dentro de 'Meu Drive > Colab Notebooks'
        target_path = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY"

        print(f"🔍 Procurando pasta em: {target_path}")

        if os.path.exists(target_path):
            print(f"✅ Sucesso! Pasta de produção localizada no diretório do Colab.")
            return target_path
        else:
            # Caso o atalho tenha sido movido, fazemos uma busca rápida
            print("⚠️ Atalho não encontrado no caminho específico. A procurar no Drive...")
            for root, dirs, files in os.walk('/content/drive/MyDrive'):
                if 'MEDIAUPSCALE_FACTORY' in dirs:
                    found = os.path.join(root, 'MEDIAUPSCALE_FACTORY')
                    print(f"✅ Pasta encontrada em: {found}")
                    return found

            # Se falhar totalmente, cria na raiz por segurança
            fallback = "/content/drive/MyDrive/MEDIAUPSCALE_FACTORY"
            os.makedirs(fallback, exist_ok=True)
            print(f"📁 Pasta não localizada. Criada nova em: {fallback}")
            return fallback

    except Exception as e:
        print(f"❌ Erro de Sistema: {e}")
        return "/content"

# Define a variável global BASE_PATH para as próximas células
BASE_PATH = setup_drive_and_factory()
````

### Older Production Pipeline (Legacy)

````python
import os
import requests
import json
import base64
import time
import re
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. CONFIGURAÇÃO DO PROJETO E MATERIAIS] ---
PROJECT_CONFIG = {
    "root_folder": "Endless_Summers_Paradise",
    "project_name": "Elite_V6_Ultra_HighRes",   # Versão focada em Resolução Máxima
    "theme": "Endless Summer - Colossal surreal water park, hyper-realistic candid photography.",
    "material_water": "Ultra-viscous glowing gummy liquid, heavy syrup physics, internal light scattering, 8k water surface detail.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides, wobbly but structural, high-fidelity surface texture.",
    "persona": "Athletic women, 1950s style, natural unedited skin, visible pores, sweat, realistic body proportions, vintage swimwear.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows, bright clear sky, cinematic exposure.",
    "aspect_ratio": "9:16",
    "scene_count": 10
}

# --- [2. SETUP DE INFRAESTRUTURA E DRIVE] ---
def setup_environment():
    try:
        if not os.path.exists('/content/drive'):
            print("🚀 A montar o Google Drive...")
            drive.mount('/content/drive')

        base_factory = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY"
        if not os.path.exists(base_factory):
            base_factory = "/content/drive/MyDrive/MEDIAUPSCALE_FACTORY"

        root_path = os.path.join(base_factory, PROJECT_CONFIG["root_folder"])
        if not os.path.exists(root_path):
            os.makedirs(root_path, exist_ok=True)

        project_dir = os.path.join(root_path, PROJECT_CONFIG["project_name"])
        if not os.path.exists(project_dir):
            os.makedirs(project_dir, exist_ok=True)
            print(f"📂 Pasta de Produção HQ Criada: {project_dir}")

        return project_dir
    except Exception as e:
        print(f"❌ Erro no Setup: {e}")
        return "/content"

# --- [3. MOTOR DE INTELIGÊNCIA - ELITE IMAGE LOCK] ---
API_KEY = userdata.get('GOOGLE_API_KEY')

def get_validated_models():
    # Fixando o modelo Gemini 3 Pro Image para evitar downgrades
    selected_brain = "models/gemini-2.5-pro"
    selected_image = "models/gemini-3-pro-image-preview"
    return selected_brain, selected_image

MODEL_BRAIN, MODEL_IMAGE = get_validated_models()

# --- [4. FUNÇÕES DE GERAÇÃO CINEMATOGRÁFICA] ---

def call_gemini_api(url, payload, max_retries=5):
    for i in range(max_retries):
        try:
            res = requests.post(url, json=payload, timeout=120).json()
            if 'candidates' in res and res['candidates']:
                return res
        except: pass
        time.sleep(2 ** i)
    return None

def generate_full_script(project_dir):
    url = f"https://generativelanguage.googleapis.com/v1beta/{MODEL_BRAIN}:generateContent?key={API_KEY}"

    # Roteiro ajustado para forçar a abertura da primeira cena
    prompt = f"""
    [ACTOR: MASTER CINEMATOGRAPHER]
    Create a 10-scene JSON script for '{PROJECT_CONFIG['theme']}'.

    STRICT CINEMATIC RULE FOR SCENE 1:
    - Scene 1 MUST be an 'Extreme Wide Shot' or 'Aerial Master View'.
    - It must show the colossal scale of the park, multiple slides, towers, and huge crowds.
    - Focus on the architecture and the 'Establishing Shot' of the paradise.
    - Subsequent scenes can then zoom into the details (sweat, pores, sliding action).

    Output ONLY valid JSON:
    {{ "episode": "{PROJECT_CONFIG['project_name']}", "storyboard": [{{ "scene_id": 1, "title": "string", "image_prompt": "string", "video_command": "string", "audio_desc": "string" }}] }}
    """

    payload = {"contents": [{"parts": [{"text": prompt}]}], "generationConfig": {"responseMimeType": "application/json"}}

    try:
        print(f"🧠 Cérebro ({MODEL_BRAIN}): Gerando roteiro com abertura em plano aberto...")
        res = call_gemini_api(url, payload)
        if not res: return None

        script_text = res['candidates'][0]['content']['parts'][0]['text']
        script = json.loads(re.sub(r'```json\n?|```', '', script_text).strip())

        path = os.path.join(project_dir, "script_hq.json")
        with open(path, "w", encoding='utf-8') as f:
            json.dump(script, f, indent=4, ensure_ascii=False)
        return script
    except Exception as e:
        print(f"❌ Erro no roteiro: {e}")
        return None

def generate_elite_image(scene_data, project_dir):
    """Gera o Master Frame com foco em escala (Cena 1) ou detalhe (restantes)."""
    url = f"https://generativelanguage.googleapis.com/v1beta/{MODEL_IMAGE}:generateContent?key={API_KEY}"

    # Se for a cena 1, injetamos comandos de "Plano Aberto"
    composition_bonus = ""
    if scene_data['scene_id'] == 1:
        composition_bonus = "WIDE ANGLE AERIAL VIEW. Establishing shot showing the whole park scale. Distant architecture, massive slides, thousands of people. Scale comparison."
    else:
        composition_bonus = "MEDIUM CLOSE-UP. Focus on the characters and action."

    elite_prompt = (
        f"9:16 VERTICAL. {composition_bonus}. {scene_data['image_prompt']}. "
        f"ULTRA-HIGH RESOLUTION. Sharp focus on distant structures. "
        f"Authentic 35mm film photography, Kodak Portra 400. "
        f"Materials: {PROJECT_CONFIG['material_water']}, {PROJECT_CONFIG['material_structures']}. "
        f"Persona: {PROJECT_CONFIG['persona']} visible in the environment. "
        f"STRICTLY NO CGI LOOK. REALISTIC PHOTOGRAPHY."
    )

    payload = {
        "contents": [{"parts": [{"text": elite_prompt}]}],
        "generationConfig": {
            "responseModalities": ["IMAGE"]
        }
    }

    try:
        scene_id = scene_data['scene_id']
        print(f"📸 Visual ({MODEL_IMAGE}): Gerando Cena {scene_id} (Foco: {'Plano Aberto' if scene_id == 1 else 'Ação/Zoom'})...")
        res = call_gemini_api(url, payload)

        if not res: return None

        parts = res['candidates'][0].get('content', {}).get('parts', [])
        img_part = next((p for p in parts if 'inlineData' in p), None)

        if not img_part: return None

        img_b64 = img_part['inlineData']['data']
        save_path = os.path.join(project_dir, f"scene_{scene_id}_ULTRA_HQ.png")

        with open(save_path, "wb") as f:
            f.write(base64.b64decode(img_b64))
        return img_b64

    except Exception as e:
        print(f"❌ Erro visual: {e}")
        return None

# --- [5. EXECUÇÃO] ---
def start_production():
    target_dir = setup_environment()
    print(f"✅ Motor Ativo: {MODEL_BRAIN} + {MODEL_IMAGE}")

    script = generate_full_script(target_dir)
    if script:
        # Gerar a Cena 1 (O Plano Aberto)
        first_scene = script['storyboard'][0]
        print(f"🚀 Produzindo a 'Establishing Shot' do Parque...")

        img_data = generate_elite_image(first_scene, target_dir)
        if img_data:
            display(Image(data=base64.b64decode(img_data)))
            print(f"\n✅ PLANO ABERTO FINALIZADO.")
            print(f"📍 Ficheiro guardado em: {target_dir}")

if __name__ == "__main__":
    start_production()
````

### Kling V3 Single Image to Video (Legacy)

````python
import requests
import time
import os
import shutil
from google.colab import drive, userdata

# 1. Configurações de Caminho e Identificadores
# Certifique-se de que o caminho existe no seu Drive
BASE_PATH = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise/Elite_V6_Ultra_HighRes"
IMAGE_FILENAME = "scene_1_ULTRA_HQ.png"
ARQUIVO_CENA = os.path.join(BASE_PATH, IMAGE_FILENAME)

# 2. Chaves de API (Configuradas no Secrets do Colab)
KLING_API_KEY = userdata.get('EVOLINK_API_KEY')
IMGBB_API_KEY = userdata.get('IMGBB_API_KEY')
BASE_URL = "https://api.evolink.ai/v1"

def upload_imgbb(caminho_arquivo):
    """ Upload para link temporário (expira em 1 hora) """
    print(f"🔄 Fazendo upload da imagem para o ImgBB...")
    try:
        with open(caminho_arquivo, "rb") as file:
            url = "https://api.imgbb.com/1/upload"
            payload = {"key": IMGBB_API_KEY, "expiration": 3600} # 1 hora para segurança
            files = {"image": file}
            response = requests.post(url, payload, files=files)
            if response.status_code == 200:
                url_direta = response.json()["data"]["url"]
                print(f"✅ Imagem disponível em: {url_direta}")
                return url_direta
            print(f"❌ Erro ImgBB: {response.text}")
    except Exception as e:
        print(f"❌ Erro no upload: {e}")
    return None

def solicitar_geracao_kling_v3(image_url):
    """ Envia a tarefa para o Kling V3 via EvoLink """
    endpoint = f"{BASE_URL}/videos/generations"
    headers = {
        "Authorization": f"Bearer {KLING_API_KEY}",
        "Content-Type": "application/json"
    }

    # Payload ajustado conforme sugestões críticas do GPT para EvoLink
    payload = {
        "model": "kling-v3-image-to-video",
        "prompt": "Cinematic masterpiece, tropical paradise beach, crystal clear water moving gently, 4k, professional sound design",
        "image_start": image_url,
        "duration": 7,      # Mantendo os 7 segundos solicitados
        "quality": "720p",   # Padrão aceito pela API
        "sound": "on"        # Ativando áudio
    }

    print(f"🚀 Enviando para Kling V3 (7s + Áudio)...")
    res = requests.post(endpoint, json=payload, headers=headers)

    if res.status_code == 200:
        task_id = res.json().get('id')
        print(f"✅ Tarefa criada! ID: {task_id}")
        return task_id
    else:
        print(f"❌ Erro ao iniciar: {res.text}")
        return None

def salvar_video_final(video_url, task_id):
    """ Baixa o vídeo e salva no Drive usando stream para evitar corrupção """
    nome_arquivo = f"kling_v3_7s_{task_id}.mp4"
    caminho_final = os.path.join(BASE_PATH, nome_arquivo)

    print(f"📥 Iniciando download do vídeo...")
    try:
        with requests.get(video_url, stream=True) as r:
            r.raise_for_status()
            with open(caminho_final, 'wb') as f:
                shutil.copyfileobj(r.raw, f)

        if os.path.exists(caminho_final):
            print(f"🎊 SUCESSO! Vídeo salvo em: {caminho_final}")
            print(f"📦 Tamanho: {os.path.getsize(caminho_final) / (1024*1024):.2f} MB")
        else:
            print("❌ Erro: O arquivo não foi gravado no disco.")
    except Exception as e:
        print(f"❌ Falha no download: {e}")

def monitorar_progresso_evolink(task_id):
    """ Monitora a tarefa no endpoint correto da EvoLink """
    # Endpoint corrigido: /videos/generations/{id}
    endpoint = f"{BASE_URL}/videos/generations/{task_id}"
    headers = {"Authorization": f"Bearer {KLING_API_KEY}"}

    print(f"⏳ Monitorando status da geração...")

    tentativas_erro = 0
    while True:
        try:
            res = requests.get(endpoint, headers=headers)

            if res.status_code == 200:
                data = res.json()
                status = data.get("status")

                print(f"[{status.upper()}]", end=" ", flush=True)

                if status == "completed":
                    # Estrutura EvoLink: data["output"]["video_url"]
                    video_url = data.get("output", {}).get("video_url")
                    if video_url:
                        print("\n✨ Geração concluída!")
                        salvar_video_final(video_url, task_id)
                        return
                    else:
                        print("\n❌ Erro: 'video_url' não encontrado no output.")
                        return

                elif status == "failed":
                    print(f"\n❌ A geração falhou. Resposta: {data}")
                    return

                time.sleep(10) # Intervalo seguro
            else:
                print(f"\n⚠️ Status Code {res.status_code}. Tentando reconectar...")
                tentativas_erro += 1
                if tentativas_erro > 5: break
                time.sleep(10)

        except Exception as e:
            print(f"\n⚠️ Erro de conexão: {e}")
            time.sleep(10)

if __name__ == "__main__":
    # 1. Garantir que o Drive está montado
    if not os.path.exists('/content/drive'):
        drive.mount('/content/drive')

    # 2. Verificar se a imagem de origem existe
    if not os.path.exists(ARQUIVO_CENA):
        print(f"❌ Erro: Imagem não encontrada em {ARQUIVO_CENA}")
    else:
        # 3. Fluxo de execução
        url_publica = upload_imgbb(ARQUIVO_CENA)
        if url_publica:
            id_da_tarefa = solicitar_geracao_kling_v3(url_publica)
            if id_da_tarefa:
                monitorar_progresso_evolink(id_da_tarefa)
````

### Improved Kling V3 Single Image to Video (Legacy)

````python
import requests
import time
import os
import shutil
from google.colab import drive, userdata

# --- CONFIGURAÇÕES ---
BASE_PATH = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise/Elite_V6_Ultra_HighRes"
IMAGE_FILENAME = "scene_1_ULTRA_HQ.png"
ARQUIVO_CENA = os.path.join(BASE_PATH, IMAGE_FILENAME)

# Definição do nome de saída baseado na imagem original
NOME_BASE = os.path.splitext(IMAGE_FILENAME)[0] # Retira a extensão (ex: scene_1_ULTRA_HQ)
NOME_FINAL_VIDEO = f"{NOME_BASE}_VIDEO.mp4"

# --- API KEYS ---
KLING_API_KEY = userdata.get('EVOLINK_API_KEY')
IMGBB_API_KEY = userdata.get('IMGBB_API_KEY')
BASE_URL = "https://api.evolink.ai/v1"

def extrair_url_video(data):
    """ Busca recursiva por URLs .mp4 no JSON da API """
    def busca_recursiva(obj):
        if isinstance(obj, str) and obj.startswith("http") and ".mp4" in obj:
            return obj
        if isinstance(dict, type(obj)):
            for v in obj.values():
                res = busca_recursiva(v)
                if res: return res
        if isinstance(list, type(obj)):
            for item in obj:
                res = busca_recursiva(item)
                if res: return res
        return None
    return busca_recursiva(data)

def upload_imgbb(caminho_arquivo):
    print(f"🔄 Fazendo upload da imagem para o ImgBB...")
    try:
        with open(caminho_arquivo, "rb") as file:
            url = "https://api.imgbb.com/1/upload"
            payload = {"key": IMGBB_API_KEY, "expiration": 3600}
            files = {"image": file}
            response = requests.post(url, payload, files=files)
            if response.status_code == 200:
                url_direta = response.json()["data"]["url"]
                print(f"✅ Imagem disponível: {url_direta}")
                return url_direta
    except Exception as e:
        print(f"❌ Erro no upload ImgBB: {e}")
    return None

def solicitar_geracao_kling_v3(image_url):
    endpoint = f"{BASE_URL}/videos/generations"
    headers = {"Authorization": f"Bearer {KLING_API_KEY}", "Content-Type": "application/json"}

    payload = {
        "model": "kling-v3-image-to-video",
        "prompt": "Cinematic masterpiece, tropical paradise beach, crystal clear water moving gently, 4k, professional sound design",
        "image_start": image_url,
        "duration": 7,
        "quality": "720p",
        "sound": "on"
    }

    print(f"🚀 Enviando para Kling V3 (7s + Áudio)...")
    try:
        res = requests.post(endpoint, json=payload, headers=headers)
        if res.status_code in [200, 201]:
            data = res.json()
            task_id = data.get('id') or data.get('task_id') or data.get('data', {}).get('id')
            print(f"✅ Tarefa criada! ID: {task_id}")
            return task_id
    except Exception as e:
        print(f"❌ Erro na solicitação: {e}")
    return None

def salvar_video_final(video_url):
    """ Salva o vídeo com nome organizado e amigável """
    os.makedirs(BASE_PATH, exist_ok=True)
    caminho_final = os.path.join(BASE_PATH, NOME_FINAL_VIDEO)

    print(f"📥 Baixando vídeo para: {NOME_FINAL_VIDEO}")
    try:
        with requests.get(video_url, stream=True, timeout=120) as r:
            r.raise_for_status()
            with open(caminho_final, 'wb') as f:
                shutil.copyfileobj(r.raw, f)

        if os.path.exists(caminho_final) and os.path.getsize(caminho_final) > 1000:
            print(f"🎊 SUCESSO! Arquivo salvo em: {caminho_final}")
    except Exception as e:
        print(f"❌ Falha no download: {e}")

def monitorar_progresso_evolink(task_id):
    endpoints = [f"{BASE_URL}/videos/generations/{task_id}", f"{BASE_URL}/tasks/{task_id}"]
    headers = {"Authorization": f"Bearer {KLING_API_KEY}"}
    idx_endpoint = 0

    print(f"⏳ Monitorando progresso...")
    while True:
        try:
            res = requests.get(endpoints[idx_endpoint], headers=headers)
            if res.status_code == 404 and idx_endpoint == 0:
                idx_endpoint = 1
                continue

            if res.status_code == 200:
                data = res.json()
                status = str(data.get("status") or data.get("data", {}).get("status") or "").lower()
                print(f"[{status.upper() or 'PROCESSANDO'}]", end=" ", flush=True)

                if status in ["completed", "succeeded", "finished"]:
                    video_url = extrair_url_video(data)
                    if video_url:
                        print("\n✨ Concluído!")
                        salvar_video_final(video_url)
                        break
                elif status in ["failed", "error"]:
                    print(f"\n❌ Erro na API: {data}")
                    break

            time.sleep(25) # Intervalo seguro para vídeos longos do Kling
        except Exception as e:
            print(f"\n⚠️ Erro no loop: {e}")
            time.sleep(20)

if __name__ == "__main__":
    if not os.path.exists('/content/drive'): drive.mount('/content/drive')

    if os.path.exists(ARQUIVO_CENA):
        url_img = upload_imgbb(ARQUIVO_CENA)
        if url_img:
            task_id = solicitar_geracao_kling_v3(url_img)
            if task_id: monitorar_progresso_evolink(task_id)
    else:
        print(f"❌ Imagem não encontrada: {ARQUIVO_CENA}")
````

### Early Full Production Pipeline (Legacy)

````python
import os
import requests
import json
import base64
import time
import re
import shutil
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. PRODUCTION FLAGS & CONFIG] ---
DRY_RUN = False
PROJECT_ROOT = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise - Production"

PROJECT_CONFIG = {
    "theme": "Endless Summer - Colossal surreal water park, hyper-realistic candid photography.",
    "material_water": "Ultra-viscous glowing gummy liquid, heavy syrup physics.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides.",
    "persona": "Athletic women, 1950s style, vintage swimwear.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows.",
    "aspect_ratio": "9:16",
    "scene_count": 10
}

# --- [NARRATIVE INTENTS] ---
NARRATIVE_DIRECTIONS = [
    "The Reveal: High altitude wide shot of the infinite park, structures stretching to the horizon.",
    "The Ascent: Character climbing a wobbly Jell-O staircase, sweat glistening under the noon sun.",
    "The Preparation: Standing at the edge of a massive syrup-filled tube, looking down.",
    "The First Drop: POV shot entering the tube, viscous liquid splashing against the 'camera' body.",
    "The Speed Loop: G-force intensity inside a semi-translucent green Jell-O tunnel.",
    "The Intermission: A surreal floating shower where water behaves like slow-motion honey.",
    "The Vertigo Edge: Standing on a platform so high the ground is blurred, only giant slides visible.",
    "The Vertical Plunge: Extreme POV of a vertical drop, camera locked to the character's chest/legs.",
    "The Splashdown: High impact into a deep pool of glowing gummy liquid, massive slow-motion crowns.",
    "The Aftermath: Character emerging from the syrup, liquid clinging to skin like a second translucent layer."
]

# --- [2. API KEYS & MODEL CONFIG] ---
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
KLING_API_KEY = userdata.get('EVOLINK_API_KEY')
IMGBB_API_KEY = userdata.get('IMGBB_API_KEY')
EVOLINK_BASE_URL = "https://api.evolink.ai/v1"

# Hardcoding models based on your specific requirements and confirmed availability
MODEL_BRAIN = "gemini-2.5-pro" # Fast and accurate for script JSON
MODEL_IMAGE = "gemini-3-pro-image-preview"

print(f"✅ Brain Selected: {MODEL_BRAIN}")
print(f"✅ Image Selected: {MODEL_IMAGE}\n")

# --- [3. INFRASTRUCTURE] ---

def setup_production_folder(title):
    sanitized_title = re.sub(r'[\w\s-]', '', title).strip().replace(' ', '_')
    timestamp = time.strftime("%Y%m%d_%H%M%S")
    folder_name = f"{sanitized_title}_{timestamp}"

    if not os.path.exists('/content/drive'):
        drive.mount('/content/drive')

    path = os.path.join(PROJECT_ROOT, folder_name)
    os.makedirs(path, exist_ok=True)
    print(f"📂 Created Production Directory: {path}")
    return path

# --- [4. ROBUST API HANDLER] ---

def call_gemini_safe(model_name, payload, description="API Call"):
    # Standard generateContent endpoint for both text and image-preview models
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model_name}:generateContent?key={GOOGLE_API_KEY}"

    for attempt in range(5):
        try:
            response = requests.post(url, json=payload, timeout=90)
            res_json = response.json()

            if 'candidates' in res_json:
                return res_json
            else:
                err_msg = res_json.get('error', {}).get('message', str(res_json))
                print(f"⚠️ {description} (Attempt {attempt+1}): {err_msg}")
        except Exception as e:
            print(f"❌ {description} Exception: {e}")

        # Exponential backoff: 1s, 2s, 4s, 8s, 16s
        time.sleep(2 ** attempt)
    return None

# --- [5. BRAIN AGENT] ---

def generate_storyboard():
    intents_str = "\n".join([f"- {d}" for d in NARRATIVE_DIRECTIONS])
    prompt = f"""
    Create a {PROJECT_CONFIG['scene_count']}-scene storyboard for '{PROJECT_CONFIG['theme']}'.

    BASE ON THESE INTENTS:
    {intents_str}

    Output ONLY valid JSON:
    {{
      "production_title": "Unique title",
      "storyboard": [
        {{
          "scene_id": 1,
          "title": "Scene Name",
          "image_prompt": "Cinematic prompt for high-end AI generator",
          "video_command": "Camera movement description"
        }}
      ]
    }}
    """
    payload = {
        "contents": [{"parts": [{"text": prompt}]}],
        "generationConfig": {"responseMimeType": "application/json"}
    }

    print(f"🧠 Brain ({MODEL_BRAIN}): Designing storyboard...")
    res = call_gemini_safe(MODEL_BRAIN, payload, "Storyboard Generation")

    if res:
        try:
            raw_text = res['candidates'][0]['content']['parts'][0]['text']
            return json.loads(re.sub(r'```json\n?|```', '', raw_text).strip())
        except Exception as e:
            print(f"❌ JSON Parsing Error: {e}")
    return None

# --- [6. IMAGE AGENT] ---

def generate_master_frame(scene_data, project_dir):
    if DRY_RUN:
        print(f"🧪 [DRY RUN] Image skip: {scene_data['title']}")
        return "dry_run.png"

    elite_prompt = (
        f"A professional cinematic 9:16 vertical photograph. {scene_data['image_prompt']}. "
        f"Water: {PROJECT_CONFIG['material_water']}. "
        f"Structure: {PROJECT_CONFIG['material_structures']}. "
        f"Lighting: {PROJECT_CONFIG['lighting']}. "
        f"Persona: {PROJECT_CONFIG['persona']}. "
        f"Hyper-realistic, 8k resolution, photorealistic."
    )

    # Payload for gemini-3-pro-image-preview
    payload = {
        "contents": [{"parts": [{"text": elite_prompt}]}],
        "generationConfig": {
            "responseModalities": ["IMAGE"]
        }
    }

    print(f"🌅 Image Agent ({MODEL_IMAGE}): Generating scene {scene_data['scene_id']}...")
    res = call_gemini_safe(MODEL_IMAGE, payload, f"Scene {scene_data['scene_id']} Image")

    if res:
        try:
            parts = res['candidates'][0]['content']['parts']
            # Looking for inlineData which contains the image bytes
            img_part = next((p for p in parts if 'inlineData' in p), None)

            if img_part:
                img_b64 = img_part['inlineData']['data']
                file_path = os.path.join(project_dir, f"scene_{scene_data['scene_id']}_master.png")
                with open(file_path, "wb") as f:
                    f.write(base64.b64decode(img_b64))
                return file_path
            else:
                print(f"❌ No image data found in response for Scene {scene_data['scene_id']}")
        except Exception as e:
            print(f"❌ Image extraction failed: {e}")
    return None

# --- [7. VIDEO AGENT] ---

def generate_video_sequence(image_path, scene_data, project_dir):
    if DRY_RUN: return "dry_run.mp4"

    # Upload to ImgBB to get a public URL for the video generator
    try:
        with open(image_path, "rb") as file:
            img_res = requests.post(
                "https://api.imgbb.com/1/upload",
                params={"key": IMGBB_API_KEY},
                files={"image": file}
            ).json()
        image_url = img_res["data"]["url"]
    except Exception as e:
        print(f"❌ ImgBB Upload Failed: {e}")
        return None

    headers = {"Authorization": f"Bearer {KLING_API_KEY}", "Content-Type": "application/json"}
    payload = {
        "model": "kling-v1", # or kling-v3 depending on your Evolink access
        "prompt": f"{scene_data['video_movement']}, heavy syrup physics, slow motion splashes, hyper-realistic",
        "image_start": image_url,
        "duration": 5
    }

    v_res = requests.post(f"{EVOLINK_BASE_URL}/videos/generations", json=payload, headers=headers).json()
    task_id = v_res.get('id') or v_res.get('data', {}).get('id')

    if not task_id:
        print(f"❌ Video task initiation failed: {v_res}")
        return None

    print(f"🚀 Video Task Started: {task_id}")
    while True:
        status_res = requests.get(f"{EVOLINK_BASE_URL}/tasks/{task_id}", headers=headers).json()
        data = status_res.get('data', status_res)
        status = str(data.get('status')).lower()
        print(f"[{status.upper()}]", end=" ", flush=True)

        if status in ["completed", "succeeded"]:
            video_url = data.get('video_url') or re.search(r'https://[\s"]+\.mp4', json.dumps(status_res)).group(0)
            save_path = os.path.join(project_dir, f"scene_{scene_data['scene_id']}_video.mp4")
            with requests.get(video_url, stream=True) as r:
                with open(save_path, 'wb') as f:
                    shutil.copyfileobj(r.raw, f)
            print(f"\n✅ Video Saved: {save_path}")
            return save_path
        elif status in ["failed", "error"]:
            print(f"\n❌ Video Task Failed: {status_res}")
            return None
        time.sleep(20)

# --- [8. EXECUTION LOOP] ---

def main():
    print("🌅 STARTING PRODUCTION PIPELINE")
    script = generate_storyboard()
    if not script:
        print("🛑 Critical Error: Brain failed to produce a script. Check API key and model availability.")
        return

    prod_dir = setup_production_folder(script['production_title'])

    for scene in script['storyboard']:
        print(f"\n🎬 Scene {scene['scene_id']}: {scene['title']}")
        img_file = generate_master_frame(scene, prod_dir)

        if img_file and not DRY_RUN:
            display(Image(filename=img_file))
            # Attempt video generation
            generate_video_sequence(img_file, scene, prod_dir)

    print("\n✅ ALL SCENES PROCESSED. Check your Drive for the master files.")

if __name__ == "__main__":
    main()
````

````python
import os
import requests
import json
import base64
import time
import re
import shutil
import subprocess
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. PRODUCTION FLAGS & CONFIG] ---
DRY_RUN = True  # SET TO FALSE TO START SPENDING API CREDITS
PROJECT_ROOT = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise - Production"

PROJECT_CONFIG = {
    "theme": "Endless Summer - Colossal surreal water park, 1950s candid photorealism.",
    "material_water": "Ultra-viscous glowing gummy liquid, heavy syrup physics.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides, NOT toys.",
    "persona": "Candid portraits of athletic women, vintage 1950s swimwear, screaming with laughter.",
    "lighting": "Hard natural sunlight, sharp 12pm shadows, 1950s film grain.",
    "aspect_ratio": "9:16",
    "scene_count": 10
}

# --- [2. API KEYS & DYNAMIC RESOLUTION] ---
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
KLING_API_KEY = userdata.get('EVOLINK_API_KEY')
IMGBB_API_KEY = userdata.get('IMGBB_API_KEY')
BASE_URL = "https://api.evolink.ai/v1"

def get_available_brains():
    url = f"https://generativelanguage.googleapis.com/v1beta/models?key={GOOGLE_API_KEY}"
    try:
        res = requests.get(url).json()
        valid = [m['name'].replace('models/', '') for m in res.get('models', []) if 'generateContent' in m.get('supportedGenerationMethods', [])]
        priority = ["gemini-3-pro-preview", "gemini-2.0-flash", "gemini-2.5-flash", "gemini-1.5-pro"]
        return [m for m in priority if m in valid] or valid
    except: return ["gemini-2.0-flash"]

AVAILABLE_BRAINS = get_available_brains()

# --- [3. CORE FUNCTIONS] ---

def call_brain_api(payload):
    for model in AVAILABLE_BRAINS:
        url = f"https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent?key={GOOGLE_API_KEY}"
        try:
            response = requests.post(url, json=payload, timeout=45)
            res = response.json()
            if 'candidates' in res: return res
        except: continue
    return None

def generate_storyboard():
    narrative_rules = """
    1. Scene 1: WIDE REVEAL - Zoom in slowly on the park. Endless slides, crowds.
    2. Scene 2: Big Toy Intro - Character prep for a specific colossal toy (e.g. 'The Vortex Fall').
    3. Scene 3: POV Action - Ultra-realistic high speed consistency in that toy.
    4. Scene 4: Surreal Loop - Impossible 'loop showers' or strange water physics.
    5. Scene 5-6: Impossible Physics - More colossal toys, showing athletic bodies and fun.
    6. Scene 7: Vertigo Prep - Character on extreme high platform.
    7. Scene 8-9: POV Plunge - scary POV angle showing legs/chest during vertical drop.
    8. Scene 10: Goodbye Zoom Out - Start on one toy, zoom out to show the infinite colossal park.
    """

    prompt = f"""
    [ROLE: MASTER STORYBOARD ARTIST]
    Generate a 10-scene storyboard for: {PROJECT_CONFIG['theme']}.
    MANDATORY SEQUENCE: {narrative_rules}

    OUTPUT FORMAT: You must return a JSON object with exactly {PROJECT_CONFIG['scene_count']} scenes in a list called 'storyboard'.
    Each scene needs: scene_id, title, image_prompt, video_movement.
    Ensure "production_title" is also in the root of the JSON.
    """

    print("✍️ Writer/Revisor Agent: Drafting full 10-scene sequence...")
    res = call_brain_api({"contents": [{"parts": [{"text": prompt}]}], "generationConfig": {"responseMimeType": "application/json"}})

    if res:
        raw_text = re.sub(r'```json\n?|```', '', res['candidates'][0]['content']['parts'][0]['text']).strip()
        data = json.loads(raw_text)
        return data[0] if isinstance(data, list) else data
    return None

# --- [4. PRODUCTION LOOP] ---

def main():
    script = generate_storyboard()
    if not script or 'storyboard' not in script:
        print("🛑 Failed to generate a valid 10-scene script.")
        return

    # Create Project Folder
    title = script.get('production_title', 'Paradise_Episode')
    prod_dir = os.path.join(PROJECT_ROOT, f"{re.sub(r'[\w]', '_', title)}_{time.strftime('%Y%m%d_%H%M%S')}")
    if not os.path.exists('/content/drive'): drive.mount('/content/drive')
    os.makedirs(prod_dir, exist_ok=True)

    # Save the JSON
    with open(os.path.join(prod_dir, "production_script.json"), "w") as f:
        json.dump(script, f, indent=4)

    print(f"📂 Folder Created: {prod_dir}")
    print(f"🎬 Storyboard contains {len(script['storyboard'])} scenes.")

    video_files = []

    for scene in script['storyboard']:
        print(f"\n✅ Processing Scene {scene['scene_id']}: {scene['title']}")
        print(f"   🎥 Camera: {scene['video_movement']}")

        if not DRY_RUN:
            # Here we would call the Image and Kling functions as before
            # For now, this loop confirms the 10 scenes exist.
            pass
        else:
            print(f"   🧪 [DRY RUN] Would generate image for: {scene['image_prompt'][:50]}...")

    print("\n✅ DEBUG SUCCESSFUL: Full 10-scene sequence is ready for production.")

if __name__ == "__main__":
    main()
````

### Restored Elite Configuration & Production Engine (Legacy)

````python
import os, requests, json, base64, time, re, shutil, subprocess
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. RESTORED ELITE CONFIGURATION] ---
PROJECT_CONFIG = {
    "root_folder": "MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise",
    "project_name": "Elite_V6_Ultra_HighRes",
    "theme": "Endless Summer - Colossal surreal water park, 1950s cinematic realism.",
    "material_water": "Ultra-viscous glowing gummy liquid, heavy syrup physics, internal light scattering, 8k water surface detail.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides, wobbly but structural, high-fidelity surface texture.",
    "persona": "Athletic women, 1950s style, natural unedited skin, visible pores, sweat, realistic body proportions, vintage swimwear.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows, bright clear sky, cinematic exposure.",
    "aspect_ratio": "9:16",
    "scene_count": 10,
    "audio_enabled": True,
    "narrative_logic": """
        1. Scene 1: WIDE REVEAL - Extreme Wide Shot (EWS). Massive scale. NO MACRO. Slow zoom in.
        2. Scene 2: Big Toy Intro - Character at entrance of a colossal named toy.
        3. Scene 3: POV Action - High-speed internal POV inside the toy.
        4. Scene 4: Surreal Loop - Impossible 'loop showers' or strange water physics.
        5. Scene 5-6: Impossible Physics - Colossal Jell-O slides wobbling under weight.
        6. Scene 7: Vertigo Prep - High platform jump/slide prep. Body POV.
        7. Scene 8-9: POV Plunge - Extreme speed POV, showing legs/chest.
        8. Scene 10: Goodbye Zoom Out - Start from toy, zoom out to show the infinite endless park scale.
    """
}

# --- [2. API KEYS] ---
KLING_API_KEY = userdata.get('EVOLINK_API_KEY')
IMGBB_API_KEY = userdata.get('IMGBB_API_KEY')
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
BASE_URL = "https://api.evolink.ai/v1"

# --- [3. RESTORED RECURSIVE HUNTER] ---
def extrair_url_video(data):
    def busca_recursiva(obj):
        if isinstance(obj, str) and obj.startswith("http") and ".mp4" in obj: return obj
        if isinstance(obj, dict):
            for v in obj.values():
                res = busca_recursiva(v)
                if res: return res
        if isinstance(obj, list):
            for item in obj:
                res = busca_recursiva(item)
                if res: return res
        return None
    return busca_recursiva(data)

def upload_imgbb(file_path):
    try:
        with open(file_path, "rb") as file:
            response = requests.post("https://api.imgbb.com/1/upload", params={"key": IMGBB_API_KEY}, files={"image": file})
            return response.json()["data"]["url"]
    except: return None

# --- [4. PRODUCTION ENGINE] ---
def run_factory(config):
    MODEL_BRAIN = "gemini-3-pro-preview"
    MODEL_IMAGE = "gemini-3-pro-image-preview"

    # PHASE 1: SCRIPTING
    print("📝 Drafting Narrative Script with Restored Materials...")
    story_prompt = f"""
    Create a {config['scene_count']}-scene JSON list.
    Theme: {config['theme']}
    Materials: {config['material_water']}, {config['material_structures']}
    Persona: {config['persona']}
    Logic: {config['narrative_logic']}
    Output JSON list where each object has: scene_id, title, image_prompt, video_movement, audio_prompt.
    """
    url_gemini = f"https://generativelanguage.googleapis.com/v1beta/models/{MODEL_BRAIN}:generateContent?key={GOOGLE_API_KEY}"
    res = requests.post(url_gemini, json={"contents": [{"parts": [{"text": story_prompt}]}], "generationConfig": {"responseMimeType": "application/json"}}).json()

    raw_text = re.sub(r'```json\n?|```', '', res['candidates'][0]['content']['parts'][0]['text']).strip()
    scenes_list = json.loads(raw_text)
    if isinstance(scenes_list, dict):
        scenes_list = next((v for v in scenes_list.values() if isinstance(v, list)), [scenes_list])

    # PHASE 2: FOLDER SETUP
    full_path = f"/content/drive/MyDrive/{config['root_folder']}/{config['project_name']}_{time.strftime('%Y%m%d_%H%M%S')}"
    if not os.path.exists('/content/drive'): drive.mount('/content/drive')
    os.makedirs(full_path, exist_ok=True)

    video_paths = []

    # PHASE 3: PRODUCTION LOOP
    for scene in scenes_list:
        s_id = scene.get('scene_id', 'X')
        print(f"\n🎬 Scene {s_id}: {scene.get('title')}")

        # IMAGE GENERATION
        img_prompt = (f"{scene.get('image_prompt')}. {config['lighting']}. "
                      f"Materials: {config['material_water']}, {config['material_structures']}. "
                      f"Persona: {config['persona']}. Aspect Ratio: {config['aspect_ratio']}.")

        img_url_api = f"https://generativelanguage.googleapis.com/v1beta/models/{MODEL_IMAGE}:generateContent?key={GOOGLE_API_KEY}"
        img_res = requests.post(img_url_api, json={"contents": [{"parts": [{"text": img_prompt}]}], "generationConfig": {"responseModalities": ["IMAGE"]}}).json()

        if 'candidates' in img_res:
            img_data = base64.b64decode(img_res['candidates'][0]['content']['parts'][0]['inlineData']['data'])
            img_file = f"{full_path}/scene_{s_id}.png"
            with open(img_file, "wb") as f: f.write(img_data)
            display(Image(filename=img_file, width=400))

            # VIDEO REQUEST
            url_img = upload_imgbb(img_file)
            if url_img:
                payload_kling = {
                    "model": "kling-v3-image-to-video",
                    "prompt": f"{scene.get('video_movement')}, {scene.get('audio_prompt')}, hyper-realistic physics",
                    "image_start": url_img, "duration": 7, "quality": "720p", "sound": "on" if config['audio_enabled'] else "off"
                }
                kv_res = requests.post(f"{BASE_URL}/videos/generations", json=payload_kling, headers={"Authorization": f"Bearer {KLING_API_KEY}"}).json()
                task_id = kv_res.get('id') or kv_res.get('data', {}).get('id')

                if task_id:
                    print(f"🚀 Task {task_id} Created. Monitoring with dual-endpoint logic...")
                    endpoints = [f"{BASE_URL}/videos/generations/{task_id}", f"{BASE_URL}/tasks/{task_id}"]
                    idx_endpoint = 0
                    while True:
                        try:
                            mon_res = requests.get(endpoints[idx_endpoint], headers={"Authorization": f"Bearer {KLING_API_KEY}"})
                            if mon_res.status_code == 404 and idx_endpoint == 0:
                                idx_endpoint = 1; continue

                            if mon_res.status_code == 200:
                                data = mon_res.json()
                                status = str(data.get("status") or data.get("data", {}).get("status") or "").lower()
                                print(f"[{status.upper() or 'WAITING'}]", end=" ", flush=True)

                                if status in ["completed", "succeeded", "finished"]:
                                    video_url = extrair_url_video(data)
                                    if video_url:
                                        v_file = f"{full_path}/scene_{s_id}_VIDEO.mp4"
                                        with requests.get(video_url, stream=True) as r:
                                            with open(v_file, 'wb') as f: shutil.copyfileobj(r.raw, f)
                                        if os.path.getsize(v_file) > 1000:
                                            video_paths.append(v_file)
                                            print(f"\n✅ Video Saved to Drive.")
                                            break
                                elif status in ["failed", "error"]:
                                    print(f"\n❌ Kling Error: {data}"); break
                            time.sleep(25)
                        except Exception as e:
                            print(f"\n⚠️ Loop Error: {e}"); time.sleep(20)

    # FINAL STITCH
    if video_paths:
        output = f"{full_path}/FINAL_MASTER.mp4"
        list_file = f"{full_path}/concat.txt"
        with open(list_file, "w") as f:
            for v in video_paths: f.write(f"file '{os.path.basename(v)}'\n")
        subprocess.run(f"ffmpeg -y -f concat -safe 0 -i {list_file} -c copy {output}", shell=True)
        print(f"\n🏆 PRODUCTION COMPLETE: {output}")

if __name__ == "__main__":
    run_factory(PROJECT_CONFIG)
````

### Main Video Generation Factory

````python
import os, requests, json, base64, time, re, shutil, subprocess
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. CONFIGURATION - LOCKED] ---
PROJECT_CONFIG = {
    "root_folder": "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise",
    "theme": "Endless Summer - Colossal, expansive surreal water park. 1950s cinematic realism.",
    "material_water": "Ultra-viscous, internally lit glowing gummy liquid, heavy syrup physics. 8k internal light scattering.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides, wobbly but structural.",
    "persona": "Athletic women, 1950s style, natural unedited skin, visible pores, sweat, realistic body proportions, vintage swimwear.",
    "crowd": "Diverse crowds of men, women, and children, all in 1950s vintage summer attire. Tiny distant figures defining the massive scale of the park.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows, bright clear sky, cinematic exposure. NO MACRO.",
    "aspect_ratio": "9:16",
    "scene_count": 7,
    "audio_enabled": True,
    "test_mode": True,
    "narrative_logic": """
        1. Scene 1: EWS - Infinite park reveal. Hundreds of tiny people (crowd) everywhere. No macro.
        2. Scene 2: Medium-Wide - Hero Persona at entrance of a colossal toy.
        3. Scene 3: POV Action - Internal slide view. Crowds visible through translucent walls.
        4. Scene 4: Wide Shot - Surreal Syrup Loops. Multiple families/crowds.
        5. Scene 5: High Platform POV - Looking down. Tiny people in the pools far below.
        6. Scene 6: POV Impact - Splashdown into a crowded syrup pool.
        7. Scene 7: Infinite Zoom Out - From Persona out to the massive, populated park horizon.
    """
}

# --- [2. API & MODEL DISCOVERY] ---
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')

def get_best_models():
    """Dynamically fetches available models and selects the best Pro and Flash versions."""
    url = f"https://generativelanguage.googleapis.com/v1beta/models?key={GOOGLE_API_KEY}"
    try:
        response = requests.get(url).json()
        available = [m['name'] for m in response['models'] if 'generateContent' in m['supportedGenerationMethods']]

        # Selection Logic: Prefer latest Pro for Brain, latest Flash for Speed/Images
        pro_models = sorted([m for m in available if 'pro' in m.lower()], reverse=True)
        flash_models = sorted([m for m in available if 'flash' in m.lower()], reverse=True)

        brain = pro_models[0] if pro_models else "models/gemini-pro"
        assistant = flash_models[0] if flash_models else "models/gemini-pro"

        return brain, assistant
    except Exception as e:
        print(f"⚠️ Model Discovery Failed: {e}. Falling back to defaults.")
        return "models/gemini-pro", "models/gemini-pro"

# --- [3. PRODUCTION ENGINE] ---
def run_factory(config):
    MODEL_BRAIN, MODEL_ASSISTANT = get_best_models()

    print(f"🛠️ MODE: {'[TEST/DRY-RUN]' if config['test_mode'] else '[LIVE PRODUCTION]'}")
    print(f"🧠 BRAIN: {MODEL_BRAIN} | ⚡ ASSISTANT: {MODEL_ASSISTANT}")

    SAFETY_SETTINGS = [
        {"category": "HARM_CATEGORY_HARASSMENT", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_HATE_SPEECH", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_SEXUALLY_EXPLICIT", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_DANGEROUS_CONTENT", "threshold": "BLOCK_NONE"}
    ]

    # PHASE 1: EPISODE NAMING
    print("🧠 Brainstorming Episode & Toy Name...")
    naming_prompt = "Give me a creative 3-word name for a colossal 1950s surreal water park toy made of Jell-O. Output ONLY the name."
    url_api = f"https://generativelanguage.googleapis.com/v1beta/{MODEL_BRAIN}:generateContent?key={GOOGLE_API_KEY}"

    try:
        name_req = {"contents": [{"parts": [{"text": naming_prompt}]}], "safetySettings": SAFETY_SETTINGS}
        name_res = requests.post(url_api, json=name_req).json()
        toy_name = name_res['candidates'][0]['content']['parts'][0]['text'].strip().replace(" ", "_").replace('"', '')
    except:
        toy_name = f"Gummy_Colossus_{int(time.time())}"

    print(f"🎬 EPISODE CLIMAX TOY: {toy_name}")

    # PHASE 2: SCRIPT GENERATION
    story_prompt = f"""
    Create a {config['scene_count']}-scene JSON list for the episode '{toy_name}'.
    Theme: {config['theme']}
    Hero Persona: {config['persona']}
    Background Crowd: {config['crowd']}
    Logic: {config['narrative_logic']}
    Output JSON list where each object has: scene_id, title, image_prompt, video_movement, audio_prompt.
    """

    script_req = {
        "contents": [{"parts": [{"text": story_prompt}]}],
        "generationConfig": {"responseMimeType": "application/json"},
        "safetySettings": SAFETY_SETTINGS
    }

    res_raw = requests.post(url_api, json=script_req).json()

    try:
        raw_text = re.sub(r'```json\n?|```', '', res_raw['candidates'][0]['content']['parts'][0]['text']).strip()
        scenes_list = json.loads(raw_text)
    except Exception as e:
        print(f"❌ Script Generation Error: {res_raw}")
        return

    # PHASE 3: FOLDER SETUP & TEST OUTPUT
    if not os.path.exists('/content/drive'): drive.mount('/content/drive')
    full_path = f"/content/drive/MyDrive/{config['root_folder']}/{toy_name}"

    for scene in scenes_list:
        s_id = scene.get('scene_id', 'X')
        print(f"\n🎬 Scene {s_id}: {scene.get('title')}")
        if config['test_mode']:
            print(f"🖼️ [PROMPT]: {scene.get('image_prompt')}")
            print(f"🎥 [MOVE]: {scene.get('video_movement')}")
            print(f"🎵 [AUDIO]: {scene.get('audio_prompt')}")

    print(f"\n✅ EPISODE '{toy_name}' READY FOR PRODUCTION.")

if __name__ == "__main__":
    run_factory(PROJECT_CONFIG)
````

### Check Project Folder Contents

````python
import os, requests, json, base64, time, re, shutil, subprocess
from google.colab import drive, userdata
from IPython.display import Image, display

# --- [1. GLOBAL CONFIGURATION] ---
PROJECT_CONFIG = {
    "root_folder": "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise",
    "project_name": "Colossal_JellO_Waterpark", # Adjusted for clearer episode naming
    "theme": "1950s cinematic realism. Colossal, surreal, impossible Jell-O water park with epic slides, thrilling rides, and vibrant water attractions. Hyper-realistic candid photography.",
    "persona": "Athletic women, 1950s style, natural unedited skin, visible pores, sweat, realistic body proportions, vintage swimwear.",
    "style": "1950s retro summer attire, modest high-waisted aesthetic.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows, bright clear sky, cinematic exposure.",
    "setting": "Tropical paradise resort, sprawling water park complex, palm trees, luxury architecture.",
    "materials": "Ultra-viscous internally lit glowing gummy liquid, heavy syrup physics, internal light scattering, 8k water surface detail. Massive semi-translucent Jell-O architectural slides, wobbly but structural, high-fidelity surface texture.",
    "park_elements": ["giant water slides", "wave pools", "lazy rivers", "water cannons", "splash pads", "themed areas", "towers", "flumes", "tunnels", "rapids", "centrifugal rides", "drop capsules"], # Added more specific elements
    "aspect_ratio": "9:16",
    "scene_count": 10, # Adjusted to 10 to match NARRATIVE_INTENTS length
    "audio_enabled": True,
    "test_mode": False, # Set to True for script review
    "negative_prompt": "low quality, blur, horizontal, 16:9, text, watermark, CGI, cartoon, drawing, illustration, macro lens, close-up (unless explicitly asked for), out of focus, deformed, ugly, bad anatomy, deformed hands, poor composition.", # Expanded negative prompts
    "attraction_types": ["Impossible Slide Loops", "Gravity Vortex", "Syrup Möbius Strip", "Hydro-Coaster", "Jell-O Flume", "Speed Slide", "Tsunami Bowl"], # Expanded attraction types
    "image_enhancements": "hyper-realistic, 8k resolution, photorealistic, authentic 35mm film photography, Kodak Portra 400."
}

# --- [2. NARRATIVE_INTENTS (Direct Guidance)] ---
NARRATIVE_INTENTS = [
    "Establishing Shot: High altitude, extreme wide view of the colossal water park. Show multiple slides, towers, and huge crowds. Focus on the architecture and vastness. The Reveal: High altitude wide shot of the infinite park, structures stretching to the horizon.",
    "Character's Ascent: Athletic woman climbing a wobbly, translucent Jell-O staircase towards a towering slide. Emphasize glistening skin and vintage swimwear under harsh sun. The Ascent: Character climbing a wobbly Jell-O staircase, sweat glistening under the noon sun.",
    "Preparation Moment: Athletic woman standing at the edge of a massive, syrup-filled water slide tube, looking down into the viscous liquid, anticipation on her face. The Preparation: Standing at the edge of a massive syrup-filled tube, looking down.",
    "First Drop POV: POV shot entering a translucent tube. Viscous glowing gummy liquid splashing against the 'camera'. Highlight the sensation of speed and the unique material. The First Drop: POV shot entering the tube, viscous liquid splashing against the 'camera' body.",
    "Speed Loop Thrill: Inside a semi-translucent Jell-O tunnel. Intense G-force. Emphasize the unique physics of the gummy liquid and the architectural marvel. The Speed Loop: G-force intensity inside a semi-translucent green Jell-O tunnel.",
    "Surreal Intermission: Character in a surreal, floating shower where water behaves like slow-motion, glowing honey. Emphasize texture and light scattering. The Intermission: A surreal floating shower where water behaves like slow-motion honey.",
    "Vertigo Edge: Athletic woman standing on an incredibly high water park platform. The ground and distant features are blurred by height, only giant slides are clearly visible. The Vertigo Edge: Standing on a platform so high the ground is blurred, only giant slides visible.",
    "Vertical Plunge POV: Extreme POV of a vertical water slide drop. Camera locked to the character's chest/legs, showing the impending impact with glowing gummy liquid below. The Vertical Plunge: Extreme POV of a vertical drop, camera locked to the character's chest/legs.",
    "Massive Splashdown: High impact splash into a deep pool of glowing gummy liquid. Slow-motion, massive crowns of liquid, and cheering crowds visible at the edges. The Splashdown: High impact into a deep pool of glowing gummy liquid, massive slow-motion crowns.",
    "Aftermath & Emergence: Athletic woman emerging from the glowing syrup pool. Liquid clinging to her skin like a second translucent layer. Natural, unedited skin details. The Aftermath: Character emerging from the syrup, liquid clinging to her skin like a second translucent layer."
]

# --- [3. CORE UTILITIES] ---
def gemini_call(prompt, api_key, model="gemini-1.5-pro", is_json=False):
    # Ensure the model name passed to the URL is just the ID, without the 'models/' prefix
    # The f-string already adds 'models/'
    model_id = model.replace('models/', '')
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model_id}:generateContent?key={api_key}"
    safety = [{"category": c, "threshold": "BLOCK_NONE"} for c in ["HARM_CATEGORY_HARASSMENT", "HARM_CATEGORY_HATE_SPEECH", "HARM_CATEGORY_SEXUALLY_EXPLICIT", "HARM_CATEGORY_DANGEROUS_CONTENT"]]
    payload = {"contents": [{"parts": [{"text": prompt}]}], "safetySettings": safety}
    if is_json: payload["generationConfig"] = {"responseMimeType": "application/json"}

    for i in range(3):
        try:
            res = requests.post(url, json=payload, timeout=60) # Increased timeout
            res.raise_for_status() # Raise an exception for HTTP errors (4xx or 5xx)
            data = res.json()
            if 'candidates' in data and data['candidates']:
                return data['candidates'][0]['content']['parts'][0]['text']
            elif 'promptFeedback' in data and data['promptFeedback'].get('blockReason'):
                print(f"API: Prompt blocked by safety settings: {data['promptFeedback']}")
                return None
            else:
                print(f"API: Response missing candidates or content: {data}")
        except requests.exceptions.RequestException as e:
            print(f"API Request error (attempt {i+1}): {e}")
        except json.JSONDecodeError as e:
            print(f"API JSON decode error (attempt {i+1}): {e}")
        except Exception as e:
            print(f"API Unexpected error in gemini_call (attempt {i+1}): {e}")
        time.sleep(2)
    return None

def call_gemini_image(prompt, api_key, model, scene_id, max_retries=3, timeout=90):
    model_id = model.replace('models/', '')
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model_id}:generateContent?key={api_key}"
    safety = [{"category": c, "threshold": "BLOCK_NONE"} for c in ["HARM_CATEGORY_HARASSMENT", "HARM_CATEGORY_HATE_SPEECH", "HARM_CATEGORY_SEXUALLY_EXPLICIT", "HARM_CATEGORY_DANGEROUS_CONTENT"]]
    payload = {
        "contents": [{"parts": [{"text": prompt}]}],
        "generationConfig": {"responseModalities": ["IMAGE"]},
        "safetySettings": safety
    }

    for attempt in range(max_retries):
        try:
            res = requests.post(url, json=payload, timeout=timeout)
            res.raise_for_status()
            data = res.json()

            if 'candidates' in data and data['candidates']:
                candidate = data['candidates'][0]
                if 'finishReason' in candidate and candidate['finishReason'] == 'NO_IMAGE':
                    print(f"❌ Gemini image generation for scene {scene_id} failed: Model returned NO_IMAGE (attempt {attempt+1}). Full response: {data}")
                    if attempt < max_retries - 1:
                        print("    Retrying...")
                        time.sleep(5 * (attempt + 1)) # Exponential backoff for retries
                        continue
                    return None

                # Check for actual image content
                if 'content' in candidate and 'parts' in candidate['content'] and candidate['content']['parts']:
                    img_part = next((p for p in candidate['content']['parts'] if 'inlineData' in p), None)
                    if img_part and 'data' in img_part['inlineData']:
                        return data # Return the full response for successful image data extraction

            # If not NO_IMAGE but still no content
            print(f"❌ Gemini image generation for scene {scene_id} failed: No valid image content in response (attempt {attempt+1}). Full response: {data}")

        except requests.exceptions.RequestException as e:
            print(f"API Request error for image generation (scene {scene_id}, attempt {attempt+1}): {e}")
        except json.JSONDecodeError as e:
            print(f"API JSON decode error for image generation (scene {scene_id}, attempt {attempt+1}): {e}")
        except Exception as e:
            print(f"API Unexpected error in call_gemini_image (scene {scene_id}, attempt {attempt+1}): {e}")

        if attempt < max_retries - 1:
            print("    Retrying...")
            time.sleep(5 * (attempt + 1)) # Exponential backoff
    return None

def extrair_url_video(data):
    """ Busca recursiva por URLs .mp4 no JSON da API """
    def busca_recursiva(obj):
        if isinstance(obj, str) and obj.startswith("http") and ".mp4" in obj:
            return obj
        if isinstance(obj, dict):
            for v in obj.values():
                res = busca_recursiva(v)
                if res: return res
        if isinstance(obj, list):
            for item in obj:
                res = busca_recursiva(item)
                if res: return res
        return None
    return busca_recursiva(data)

def get_best_models(api_key):
    """Dynamically fetches available models and selects the best Pro and Flash versions."""
    url = f"https://generativelanguage.googleapis.com/v1beta/models?key={api_key}"
    try:
        response = requests.get(url).json()
        available = [m['name'] for m in response['models'] if 'generateContent' in m['supportedGenerationMethods']]

        # Selection Logic: Prefer latest Pro for Brain, then specific image model, then latest Flash
        pro_models = sorted([m for m in available if 'pro' in m.lower()], reverse=True)
        flash_models = sorted([m for m in available if 'flash' in m.lower()], reverse=True)

        brain = pro_models[0] if pro_models else "models/gemini-pro"

        # Prioritize 'gemini-3-pro-image-preview' for assistant if available
        if "models/gemini-3-pro-image-preview" in available:
            assistant = "models/gemini-3-pro-image-preview"
        elif flash_models:
            assistant = flash_models[0]
        else:
            assistant = "models/gemini-pro"

        return brain, assistant
    except Exception as e:
        print(f"⚠️ Model Discovery Failed: {e}. Falling back to defaults.")
        return "models/gemini-pro", "models/gemini-pro"

# --- [4. PRODUCTION FACTORY] ---
def run_factory(cfg, narrative_intents):
    G_KEY, K_KEY, I_KEY = userdata.get('GOOGLE_API_KEY'), userdata.get('EVOLINK_API_KEY'), userdata.get('IMGBB_API_KEY')
    B_URL, M_VID = "https://api.evolink.ai/v1", "kling-v3-image-to-video"

    # Get dynamic models
    MODEL_BRAIN_SELECTED, MODEL_ASSISTANT_SELECTED = get_best_models(G_KEY)

    print(f"🛠️ MODE: {'[TEST/DRY-RUN]' if cfg['test_mode'] else '[LIVE PRODUCTION]'}")
    print(f"🧠 BRAIN: {MODEL_BRAIN_SELECTED} | ⚡ ASSISTANT: {MODEL_ASSISTANT_SELECTED}")

    # 1. NAME & FOLDER
    raw_n = gemini_call(f"3-word name for a water park ride from: {cfg['attraction_types']}. Output ONLY the name.", G_KEY, model=MODEL_BRAIN_SELECTED)
    ride_name = re.sub(r'\W+', '_', (raw_n or "Jello_Resort").strip())
    # Combine the project_name from config and the generated ride name for the final folder/project name
    project_folder_name = f"{cfg['project_name']}_{ride_name}"

    if not os.path.exists('/content/drive'): drive.mount('/content/drive')
    path = f"/content/drive/MyDrive/{cfg['root_folder']}/{project_folder_name}"
    if not cfg['test_mode']: os.makedirs(path, exist_ok=True)
    print(f"🚀 PRODUCTION: {project_folder_name}")

    # 2. SCRIPTING
    print("📝 Generating script...")
    narrative_intents_str = "\n".join([f"- Scene {i+1}: {intent}" for i, intent in enumerate(narrative_intents)])

    script_prompt = f"""
As a Master Storyboard Artist, create a {cfg['scene_count']}-scene JSON storyboard for a '{cfg['theme']}' episode, featuring a '{ride_name}' ride.
Each scene's description MUST be hyper-realistic and include specific details from '{cfg['materials']}' and '{cfg['persona']}'.

Follow these narrative intents for each scene:
{narrative_intents_str}

Output JSON list where each object has: 'scene_id', 'title', 'image_logic' (detailed visual description), 'motion_logic' (detailed camera movement), 'audio_prompt' (sound design).
"""

    raw_j = gemini_call(script_prompt, G_KEY, is_json=True, model=MODEL_BRAIN_SELECTED)
    if not raw_j: return print("❌ Script Failure: Gemini did not return a valid script.")

    try:
        # Clean the raw JSON string from potential markdown wrappers or other artifacts
        cleaned_json_str = re.sub(r'```json\n?|```', '', raw_j).strip()
        scenes = json.loads(cleaned_json_str)

        # Ensure scenes is a list and assign scene_id if missing or malformed
        if not isinstance(scenes, list):
            print("⚠️ Warning: Script did not return a list. Attempting to convert.")
            scenes = [scenes] # Wrap in a list if it's a single object

        for i, scene in enumerate(scenes):
            if 'scene_id' not in scene or not isinstance(scene['scene_id'], (int, str)):
                scene['scene_id'] = i + 1
            else:
                try:
                    scene['scene_id'] = int(scene['scene_id'])
                except (ValueError, TypeError):
                    scene['scene_id'] = i + 1 # Fallback if conversion fails

    except json.JSONDecodeError as e:
        print(f"❌ JSON Parsing Error: {e}")
        print(f"Raw JSON string received: {raw_j}")
        return
    except Exception as e:
        print(f"❌ Unexpected Error during JSON parsing: {e}")
        return

    # Save the generated JSON script to the production folder
    if not cfg['test_mode']:
        script_file_path = os.path.join(path, "production_script.json")
        with open(script_file_path, "w", encoding='utf-8') as f:
            json.dump(scenes, f, indent=4, ensure_ascii=False)
        print(f"✅ Production script saved to: {script_file_path}")

    video_paths = []

    for scene in scenes:
        s_id = scene.get('scene_id', 'X')
        print(f"\n🎬 Scene {s_id}: {scene.get('title')}")

        if not cfg['test_mode']:
            print(f"🖼️ Image Generation Model: {MODEL_ASSISTANT_SELECTED}") # Print image model
            # IMAGE GENERATION
            img_prompt = (
                f"'{scene.get('image_logic')}.' "  # The AI-generated image description for the scene
                f"Theme: {cfg['theme']}. "
                f"Lighting: {cfg['lighting']}. "
                f"Materials: {cfg['materials']}. "
                f"Persona: {cfg['persona']}. "
                f"Aspect Ratio: {cfg['aspect_ratio']}. "
                f"{cfg['image_enhancements']}. "
                f"Negative prompt: {cfg['negative_prompt']}"
            )

            # Use the new helper function for image generation
            img_res = call_gemini_image(img_prompt, G_KEY, MODEL_ASSISTANT_SELECTED, s_id)

            if img_res: # Check if img_res is not None (meaning successful call_gemini_image)
                candidate = img_res['candidates'][0]
                img_part = next((p for p in candidate['content']['parts'] if 'inlineData' in p), None)
                img_data = base64.b64decode(img_part['inlineData']['data'])
                img_file = f"{path}/scene_{s_id}.png"
                with open(img_file, "wb") as f: f.write(img_data)
                display(Image(filename=img_file, width=400))

                # VIDEO REQUEST
                print(f"⬆️ Uploading image for scene {s_id} to ImgBB...")
                imgbb_url = "https://api.imgbb.com/1/upload"
                imgbb_files = {'image': open(img_file, 'rb')}
                imgbb_payload = {'key': I_KEY, 'expiration': 3600}
                try:
                    imgbb_res = requests.post(imgbb_url, files=imgbb_files, data=imgbb_payload).json()
                    url_img = imgbb_res["data"]["url"]
                    print(f"✅ ImgBB URL obtained for scene {s_id}.")
                except Exception as e:
                    print(f"❌ ImgBB upload failed for scene {s_id}: {e}")
                    url_img = None

                if url_img:
                    payload_kling = {
                        "model": M_VID,
                        "prompt": f"{scene.get('motion_logic')}, {scene.get('audio_prompt')}, hyper-realistic physics, {cfg['materials']}, {cfg['persona']}, {cfg['style']}",
                        "image_start": url_img, "duration": 7, "quality": "720p", "sound": "on" if cfg['audio_enabled'] else "off"
                    }
                    kling_headers = {"Authorization": f"Bearer {K_KEY}", "Content-Type": "application/json"}
                    kv_res = requests.post(f"{B_URL}/videos/generations", json=payload_kling, headers=kling_headers).json()
                    task_id = kv_res.get('id') or kv_res.get('data', {}).get('id')

                    if task_id:
                        print(f"🚀 Kling Task {task_id} Created for scene {s_id}. Monitoring...")
                        endpoints = [f"{B_URL}/videos/generations/{task_id}", f"{B_URL}/tasks/{task_id}"]
                        idx_endpoint = 0
                        while True:
                            try:
                                mon_res = requests.get(endpoints[idx_endpoint], headers=kling_headers)
                                if mon_res.status_code == 404 and idx_endpoint == 0:
                                    idx_endpoint = 1; continue

                                if mon_res.status_code == 200:
                                    data = mon_res.json()
                                    status = str(data.get("status") or data.get("data", {}).get("status") or "").lower()
                                    print(f"[{status.upper() or 'WAITING'}]", end=" ", flush=True)

                                    if status in ["completed", "succeeded", "finished"]:
                                        video_url = extrair_url_video(data)
                                        if video_url:
                                            v_file = f"{path}/scene_{s_id}_VIDEO.mp4"
                                            print(f"\n📥 Downloading video for scene {s_id} from {video_url}")
                                            with requests.get(video_url, stream=True, timeout=120) as r:
                                                r.raise_for_status() # Raise HTTPError for bad responses (4xx or 5xx)
                                                with open(v_file, 'wb') as f: shutil.copyfileobj(r.raw, f)
                                            if os.path.exists(v_file) and os.path.getsize(v_file) > 1000:
                                                video_paths.append(v_file)
                                                print(f"✅ Video for scene {s_id} saved to Drive.")
                                                break
                                            else:
                                                print(f"❌ Downloaded video for scene {s_id} is too small or corrupted.")
                                                break
                                        else:
                                            print(f"\n❌ Video URL not found in response for scene {s_id}.")
                                            break
                                    elif status in ["failed", "error"]:
                                        print(f"\n❌ Kling Error for scene {s_id}: {data}"); break
                                time.sleep(25)
                            except requests.exceptions.RequestException as e:
                                print(f"\n⚠️ Network/API Error while monitoring scene {s_id}: {e}"); time.sleep(20)
                            except Exception as e:
                                print(f"\n⚠️ Unexpected Error while monitoring scene {s_id}: {e}"); time.sleep(20)
                    else:
                        print(f"❌ Failed to get Kling task ID for scene {s_id}: {kv_res}")
                else:
                    print(f"❌ Skipping video generation for scene {s_id} due to ImgBB upload failure.")
            else:
                print(f"❌ Image generation failed for scene {s_id} after retries.")
        else:
            print(f"   🖼️ [PROMPT]: {scene.get('image_logic')}")
            print(f"   🎥 [MOVE]: {scene.get('motion_logic')}")
            print(f"   🎵 [AUDIO]: {scene.get('audio_prompt')}")

    # FINAL STITCH
    if not cfg['test_mode'] and video_paths:
        output_filename = f"{project_folder_name}_MASTER.mp4"
        output = os.path.join(path, output_filename)
        list_file = os.path.join(path, "concat.txt")
        with open(list_file, "w") as f:
            for v in video_paths: f.write(f"file '{os.path.basename(v)}'\n")
        try:
            print(f"\n✨ Concatenating {len(video_paths)} videos into {output_filename}...")
            # Changed from -c copy to re-encode for robustness
            subprocess.run(f"ffmpeg -y -f concat -safe 0 -i {list_file} -c:v libx264 -preset fast -crf 23 -c:a aac {output}", shell=True, check=True)
            print(f"\n🏆 PRODUCTION COMPLETE: {output}")
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg concatenation failed: {e}")
            print(f"FFmpeg stderr: {e.stderr.decode()}")
        except FileNotFoundError:
            print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.")
    elif cfg['test_mode']:
        print(f"\n✅ EPISODE '{project_folder_name}' READY FOR PRODUCTION (TEST MODE).")
    else:
        print(f"\n⚠️ No videos were generated for concatenation.")

if __name__ == "__main__":
    run_factory(PROJECT_CONFIG, NARRATIVE_INTENTS)
````

### Function: Merge Individual Scene Videos into Master

````python
import os
import subprocess
import re
from google.colab import drive

def merge_scene_videos(project_folder_path):
    """
    Merges all '_VIDEO.mp4' files found in the given project_folder_path
    into a single master MP4 file, sorted by scene ID.

    Args:
        project_folder_path (str): The absolute path to the project folder
                                   containing the individual scene videos.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...")
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}")
        return

    print(f"🎥 Initiating video merge for project: {os.path.basename(project_folder_path)}")

    # 1. Collect and sort video files
    all_files = os.listdir(project_folder_path)
    raw_video_files = [f for f in all_files if f.endswith('_VIDEO.mp4')]

    if not raw_video_files:
        print(f"⚠️ No individual '_VIDEO.mp4' files found in {project_folder_path}. Nothing to merge.")
        return

    # Sort files numerically by extracting the scene ID
    def get_scene_id(filename):
        match = re.search(r'scene_(\d+)_VIDEO\.mp4', filename)
        return int(match.group(1)) if match else 0 # Default to 0 if ID not found

    video_files = sorted(raw_video_files, key=get_scene_id)

    # 2. Prepare the concat.txt file
    list_file = os.path.join(project_folder_path, "concat.txt")
    with open(list_file, "w") as f:
        for v in video_files: f.write(f"file '{v}'\n") # Write relative paths since cwd will be set
    print(f"✅ Generated {list_file} with {len(video_files)} video entries.")

    # 3. Define output filename and path
    output_filename = f"{os.path.basename(project_folder_path)}_MASTER.mp4"
    output_file = os.path.join(project_folder_path, output_filename)

    # 4. Execute FFmpeg command for concatenation and re-encoding
    try:
        print(f"\n✨ Concatenating {len(video_files)} videos into {output_filename}...")
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_file}\"",
            shell=True,
            check=True,
            cwd=project_folder_path, # IMPORTANT: Run ffmpeg from the project directory
            capture_output=True # Capture stdout and stderr for debugging
        )

        if os.path.exists(output_file) and os.path.getsize(output_file) > 1000:
            print(f"\n🏆 MERGE COMPLETE! Master video saved to: {output_file}")
        else:
            print(f"❌ Merged video '{output_filename}' either not created or is too small/corrupted.")

    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg concatenation failed: {e}")
        print(f"FFmpeg stdout: {e.stdout.decode('utf-8')}")
        print(f"FFmpeg stderr: {e.stderr.decode('utf-8')}")
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.")
    except Exception as e:
        print(f"❌ An unexpected error occurred during merging: {e}")


# --- Example Usage (Call this function after running this cell) ---
# Define the base project root (adjust if your structure changes)
project_root_config = "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise"
base_path = f"/content/drive/MyDrive/{project_root_config}"

# The specific project folder you want to concatenate
folder_to_merge = "Technicolor_Gravity_Dream_20260416_050338"
full_path_to_merge = os.path.join(base_path, folder_to_merge)

print(f"\nAttempting to merge for: {full_path_to_merge}\n")
merge_scene_videos(full_path_to_merge)
````

````python
import os, requests, json, base64, time, re, shutil, subprocess
from google.colab import drive, userdata
from IPython.display import Image, display
from elevenlabs.client import ElevenLabs

# --- [1. GLOBAL CONFIGURATION - REFINED] ---
PROJECT_CONFIG = {
    "root_folder": "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise - Production",
    "theme": "1950s cinematic realism. Colossal, surreal, impossible, dream-like, futuristic Jell-O water park with epic slides, thrilling rides, and vibrant water attractions. Hyper-realistic candid photography, capturing pure joy and exhilaration.",
    "persona": "Beautiful, attractive, striking beauty, model body athletic women, 1950s style, natural unedited skin, visible pores, sweat, realistic body proportions, wearing diverse vintage swimwear/wetsuits in random, vibrant colors and patterns, laughing and screaming with pure joy. NO SHOES, NO FOOTWEAR.",
    "style": "vintage color palettes, retro summer attire, modest high-waisted aesthetic, with all people in the park wearing beautiful vintage suits in vibrant and diverse colors.",
    "lighting": "Hard natural sunlight, 12pm high-noon shadows, bright clear sky, cinematic exposure.",
    "setting": "Tropical paradise resort, sprawling futuristic water park complex, gleaming structures, lush foliage.",
    "material_water": "Ultra-viscous, internally lit glowing gummy liquid, heavy syrup physics, internal light scattering, 8k water surface detail. Gelatina, Ultra-viscous candy dream like materials. IT MUST BE JELLO, NOT WATER.",
    "material_structures": "Massive semi-translucent Jell-O architectural slides, wobbly but structurally sound, high-fidelity surface texture, appearing impossibly strong yet flexible. Gelatina, Ultra-viscous candy dream like materials. IT MUST BE JELLO, NOT WATER.",
    "park_elements": ["giant water slides", "wave pools", "lazy rivers", "water cannons", "splash pads", "themed areas", "towers", "flumes", "tunnels", "rapids", "centrifugal rides", "drop capsules", "sky-high platforms", "anti-gravity zones", "liquid portals"],
    "aspect_ratio": "9:16",
    "scene_count": 9, # Adjusted to 9 as Scene 9 is removed
    "audio_enabled": False, # Audio turned OFF for Kling as per user clarification
    "elevenlabs_sfx_enabled": True, # New flag for ElevenLabs sound effects
    "test_mode": False, # Set to True for script review
    "negative_prompt": "low quality, blur, horizontal, wide-angle distortion, stretched, warped, miniature, macro photography, toy scale, tabletop, model kit, tilt-shift, blurry foreground, kitchen background, laboratory, indoor, low-ceiling, small scale, low resolution, watermark, text, 16:9 ratio, NO TEXT OVERLAY, NO WATERMARKS, NO DATE STAMPS, NO METADATA, NO BRANDING, NO LOGOS, NO COPYRIGHT SYMBOLS., digital art, sketch, painting, drawing, cartoon, anime, 3D render, bad art, low-res, pixelated, abstract, surreal, impressionistic, graphic, out of frame, cropped, distorted, ugly, tiling, fake, generated, AI, neural network, render, software, bad photo, amateur photo, split image, multiple frames, grid, mosaic, triptych, diptych, collage, no split screen, NO FLOATING PEOPLE, NO FLYING OBJECTS, NO TEXT IN IMAGES, NO BRANDING, NO LOGOS, NO COPYRIGHT SYMBOLS., shoes, footwear.",
    "attraction_types": ["Impossible Slide Loops", "Gravity Vortex", "Syrup Möbius Strip", "Hydro-Coaster", "Jell-O Flume", "Speed Slide", "Tsunami Bowl", "Sky-Drop Capsule", "Dream-Stream Flow", "Zero-G Water Tube", "Nebula Falls"],
    "image_enhancements": "hyper-realistic, 8k resolution, photorealistic, authentic 35mm film photography style (no text or branding), Kodak Portra 400 aesthetic, pure joy, exhilarating, portrait composition, vertical framing, 9:16 vertical aspect ratio, indistinguishable from reality, cinematic quality, professional photography, not AI generated, zero AI look, single shot, one single image, single composition, no diptych, no triptych, no collage, no split screen.",
    "music_prompt_base": "1950s cinematic, retro vintage summer resort orchestral piece, with a touch of mystery and awe, slightly eerie from a distant old speaker. Emphasize strings and brass, no vocals."
}

# --- [2. NARRATIVE_INTENTS - REFINED] ---
NARRATIVE_INTENTS = [
    "Establishing Shot: Extreme high altitude, dynamic aerial wide angle view, looking forward and slightly down, of the colossal, futuristic water park. Emphasize the long, winding, and vertically oriented visual flow of the slides and impossible dream-like toys. Zoom in slowly, getting closer to the toys and showing people having lots of fun, also people scarrying and screaming of joy. Focus on architectural grandeur and vastness, creating a sense of awe and slight vertigo from height. Must strictly adhere to the project's defined hard natural sunlight, 12pm high-noon shadows, and bright clear sky.",
    "Image Logic (for image_gen): WIDE ANGLE LENS. Extreme split-depth (over-under) shot. Beneath the water line: Back view of a tanned, athletic woman (persona) in a floral 1950s swimwear (NO SHOES, NO FOOTWEAR), gracefully swimming towards the translucent slide. Her legs are visible, propelling her through the ultra-viscous glowing gummy liquid, with bubbles and caustic light refraction on her skin. She MUST NOT disappear from the scene, her form clear. Above the water line: A colossal, white Googie-style architectural water slide made of {cfg['material_structures']} and other huge, impossible toy structures are clearly visible in the background, showing the vastness of the park. High-noon hard sunlight, 1950s cinematic film stock, hyper-realistic water surface tension. Motion Logic (for kling): Smooth, slow-motion tracking shot from behind the swimming woman, keeping her centered as she approaches and enters the translucent slide opening. Her legs continue to propel her. The camera follows her into the channel, maintaining focus on her as she enters the slide. Cinematic water turbulence and bubble trails are evident. Audio Prompt: Sound of a woman swimming, gentle splashes, muffled happy screams and laughter of people in the background, distant 1950s summer resort ambiance, continuous light splashing and murmuring crowd noise.",
    "High Speed POV Drop (Continuation from Scene 2): Direct, immersive POV sequence from *inside* the same high slide or toy from Scene 2. The viewer feels they are actually entering the toy. **Extremely dynamic, high-speed plunge from a great height**, with the viscous glowing gummy liquid rushing past and creating motion blur. The view is from *high up*, showing the tanned legs and bare feet (NO SHOES, NO FOOTWEAR) of the persona in vintage swimwear, with a strong, dynamic blur of futuristic park elements and distant screaming people through the transparent walls. Emphasize **visceral, rapid descent** and exhilarating vertigo, an extreme physics fall with a strong sense of *moving through the liquid*. Must ensure authentic 35mm film photography, Kodak Portra 400 aesthetic. Maintain hard natural sunlight and bright clear sky. Audio Prompt: Intense rushing water sounds, whooshing wind, woman's exhilarating scream, distant laughter of children and adults, constant background water park ambient noise.",
    "Impossible Attraction 1: Introduce a new, unique, 'dream-like' impossible water park attraction (e.g., 'Syrup Möbius Strip' or 'Nebula Falls'). The camera is positioned from down in the {cfg['material_water']}, capturing the vintage models in their beautiful swimwear (NO SHOES, NO FOOTWEAR) with the surreal toy in the background. Show people laughing and screaming with joy as they experience its surreal physics and playful design. Focus on the creative and fantastical aspects of this specific ride, distinct from others. Maintain hard natural sunlight and bright clear sky. **STRICTLY NO TEXT, NUMBERS, OR LOGOS IN THE IMAGE.** Audio Prompt: Joyful screams and laughter from women and children, loud gushing and sloshing of viscous liquid, playful splashing, upbeat 1950s resort ambiance.",
    "Impossible Attraction 2: Another new, 'futuristic' impossible water park attraction (e.g., 'Zero-G Water Tube' or 'Liquid Portal'). Capture the bizarre, gravity-defying, or visually stunning aspects. People (persona) in vintage swimwear (NO SHOES, NO FOOTWEAR) are actively engaged, expressing pure exhilaration. Emphasize the impossible and highly imaginative nature of the toy. *Ensure a Kodak Portra 400 film aesthetic without any text overlays on the image.* Maintain hard natural sunlight and bright clear sky. Audio Prompt: Sounds of awe and wonder from people, unique and strange gurgling liquid sounds, occasional excited shrieks, constant background chatter and water park sounds.",
    "Impossible Attraction 3: A third unique, surreal water park feature (e.g., 'Dream-Stream Flow' or 'Crystalline Shower Loops'). Highlight its fantastical design and the joyful, playful interactions of the crowd. Ensure it looks distinct and enhances the 'dream-like' and 'futuristic' theme, with happy participants. Maintain hard natural sunlight and bright clear sky. Audio Prompt: Giggling and happy exclamations from women, smooth flowing and bubbling liquid sounds, light splashes, continuous cheerful background noise of a busy water park.",
    "Image Logic (for image_gen): Extremely high-angle aerial drone shot looking down from a stratospheric platform high above the clouds. At the edge of the platform sits a massive, chrome-and-neon 'Quantum Leap' entrance portal tower (as seen in). A lone athletic woman (persona) in vintage swimwear (NO SHOES, NO FOOTWEAR) walks toward the glowing blue interface, which reads 'ACCESS GRANTED' and shows a countdown timer '03:14'. Below, a dizzying vertical drop into a surreal tropical Jell-O park far below. Harsh tropical sun creating sharp shadows. 1950s cinematic film stock, telephoto compression to maximize vertigo. Motion Logic (for kling): Tracking shot behind the model as they approach the portal. They grip the side handles (like in) and push off. Sudden forward acceleration and camera shake. The light changes instantly from bright sun to internal blue/purple neon streaks as they enter the pitch-black tube. Audio Prompt: Woman's confident footsteps, electronic beeps and whooshes from the portal, a whoosh as she pushes off, followed by a rapid whoosh and high-pitched excited yell, futuristic liquid sounds, distant water park ambience.",
    "Image Logic (for image_gen): **Extreme High-Altitude POV**, from the exact perspective of the persona sitting at the very top of an impossible, colossal vertical slide drop. **Visible are tanned legs and bare feet (NO SHOES, NO FOOTWEAR) dangling directly over a terrifying {cfg['material_structures']} abyss, with the *entire park visible far below*, stretching to the horizon, creating an intense sense of vertigo and height. The feet are positioned to fall *into* the slide, not out.** Harsh high-noon 12pm sun glints off the skin. The slide itself is translucent {cfg['material_structures']} looping down towards an ocean horizon. Hyper-realistic texture on skin and concrete, ultra-vibrant Jell-O colors, massive depth of field. Motion Logic (for kling): **Heart-stopping vertical free-fall drop, *starting from this extreme high vantage point with the park below*.** Rapid camera acceleration as the subject slides down. Handheld-style vibration. The park below zooms in with violent, frame-shaking speed. Extreme motion blur on the slide walls while the feet and legs remain in sharp focus, G-force effect on the camera lens simulation. **The viewer must feel the immense height and the impending fall.** Audio Prompt: Loud, building wind rush, a high-pitched scream from the woman turning into a yell of pure joy, then a powerful, deep whoosh of rushing liquid, distant echoing screams of others, a constant sense of height and falling.",
    "Grand Zoom Out: Begin with a wide shot of a particularly vibrant and busy section of the futuristic water park, filled with people (persona and crowds) laughing, screaming, and playing on an impossible attraction. Slowly and majestically zoom out, revealing more and more of the endless, sprawling, dream-like water park complex until it becomes a breathtaking, vast tapestry of color and motion, showcasing the full scale and endless summer joy. Must maintain a strict vertical composition (9:16 aspect ratio). Must ensure that all park elements are visible within the frame, especially the long, winding, and vertically oriented visual flow of the slides and impossible toys. Maintain hard natural sunlight and bright clear sky. Audio Prompt: A rich tapestry of distant park sounds: thousands of people screaming and laughing, a cacophony of splashes and gushing liquid, all blending into a harmonious, joyful symphony, punctuated by a powerful, uplifting orchestral swell of the retro vintage music from PROJECT_CONFIG['music_prompt_base'] as the camera zooms out."
]

# --- [3. CORE UTILITIES - COPIED FOR SELF-CONTAINMENT] ---
def gemini_call(prompt, api_key, model="gemini-1.5-pro", is_json=False):
    model_id = model.replace('models/', '')
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model_id}:generateContent?key={api_key}"
    safety = [{"category": c, "threshold": "BLOCK_NONE"} for c in ["HARM_CATEGORY_HARASSMENT", "HARM_CATEGORY_HATE_SPEECH", "HARM_CATEGORY_SEXUALLY_EXPLICIT", "HARM_CATEGORY_DANGEROUS_CONTENT"]]
    payload = {"contents": [{"parts": [{"text": prompt}]}], "safetySettings": safety}
    if is_json: payload["generationConfig"] = {"responseMimeType": "application/json"}

    for i in range(3):
        try:
            res = requests.post(url, json=payload, timeout=60)
            res.raise_for_status()
            data = res.json()
            if 'candidates' in data and data['candidates']:
                return data['candidates'][0]['content']['parts'][0]['text']
            elif 'promptFeedback' in data and data['promptFeedback'].get('blockReason'):
                print(f"API: Prompt blocked by safety settings: {data['promptFeedback']}", flush=True)
                return None
            else:
                print(f"API: Response missing candidates or content: {data}", flush=True)
        except requests.exceptions.RequestException as e:
            print(f"API Request error (attempt {i+1}): {e}", flush=True)
        except json.JSONDecodeError as e:
            print(f"API JSON decode error (attempt {i+1}): {e}", flush=True)
        except Exception as e:
            print(f"API Unexpected error in gemini_call (attempt {i+1}): {e}", flush=True)
        time.sleep(2)
    return None

def call_gemini_image(prompt, api_key, model, scene_id, max_retries=3, timeout=90):
    model_id = model.replace('models/', '')
    url = f"https://generativelanguage.googleapis.com/v1beta/models/{model_id}:generateContent?key={api_key}"
    safety = [{"category": c, "threshold": "BLOCK_NONE"} for c in ["HARM_CATEGORY_HARASSMENT", "HARM_CATEGORY_HATE_SPEECH", "HARM_CATEGORY_SEXUALLY_EXPLICIT", "HARM_CATEGORY_DANGEROUS_CONTENT"]]
    payload = {
        "contents": [{"parts": [{"text": prompt}]}],
        "generationConfig": {"responseModalities": ["IMAGE"]},
        "safetySettings": safety
    }

    for attempt in range(max_retries):
        try:
            res = requests.post(url, json=payload, timeout=timeout)
            res.raise_for_status()
            data = res.json()

            if 'candidates' in data and data['candidates']:
                candidate = data['candidates'][0]
                if 'finishReason' in candidate and candidate['finishReason'] == 'NO_IMAGE':
                    print(f"❌ Gemini image generation for scene {scene_id} failed: Model returned NO_IMAGE (attempt {attempt+1}). Full response: {data}", flush=True)
                    if attempt < max_retries - 1:
                        print("    Retrying:", flush=True)
                        time.sleep(5 * (attempt + 1))
                        continue
                    return None

                if 'content' in candidate and 'parts' in candidate['content'] and candidate['content']['parts']:
                    img_part = next((p for p in candidate['content']['parts'] if 'inlineData' in p), None)
                    if img_part and 'data' in img_part['inlineData']:
                        return data

            print(f"❌ Gemini image generation for scene {scene_id} failed: No valid image content in response (attempt {attempt+1}). Full response: {data})", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"API Request error for image generation (scene {scene_id}, attempt {attempt+1}): {e}", flush=True)
        except json.JSONDecodeError as e:
            print(f"API JSON decode error for image generation (scene {scene_id}, attempt {attempt+1}): {e}", flush=True)
        except Exception as e:
            print(f"API Unexpected error in call_gemini_image (scene {scene_id}, attempt {attempt+1}): {e}", flush=True)

        if attempt < max_retries - 1:
            print("    Retrying:", flush=True)
            time.sleep(5 * (attempt + 1))
    return None

def extrair_url_video(data):
    def busca_recursiva(obj):
        if isinstance(obj, str) and obj.startswith("http") and ".mp4" in obj: return obj
        if isinstance(obj, dict):
            for v in obj.values():
                res = busca_recursiva(v)
                if res: return res
        if isinstance(obj, list):
            for item in obj:
                res = busca_recursiva(item)
                if res: return res
        return None
    return busca_recursiva(data)

def get_best_models(api_key):
    url = f"https://generativelanguage.googleapis.com/v1beta/models?key={api_key}"
    try:
        response = requests.get(url).json()
        available = [m['name'] for m in response['models'] if 'generateContent' in m['supportedGenerationMethods']]

        pro_models = sorted([m for m in available if 'pro' in m.lower()], reverse=True)
        flash_models = sorted([m for m in available if 'flash' in m.lower()], reverse=True)

        brain = pro_models[0] if pro_models else "models/gemini-pro"

        if "models/gemini-3-pro-image-preview" in available:
            assistant = "models/gemini-3-pro-image-preview"
        elif flash_models:
            assistant = flash_models[0]
        else:
            assistant = "models/gemini-pro"

        return brain, assistant
    except Exception as e:
        print(f"⚠️ Model Discovery Failed: {e}. Falling back to defaults.", flush=True)
        return "models/gemini-pro", "models/gemini-pro"

# --- [4. PRODUCTION FACTORY] ---
def run_factory(config):
    G_KEY, K_KEY, I_KEY = userdata.get('GOOGLE_API_KEY'), userdata.get('EVOLINK_API_KEY'), userdata.get('IMGBB_API_KEY')
    B_URL, M_VID = "https://api.evolink.ai/v1", "kling-v3-image-to-video"

    # Nested helper functions for audio processing
    def get_video_duration(video_path):
        try:
            cmd = [
                "ffprobe", "-v", "error",
                "-show_entries", "format=duration",
                "-of", "default=noprint_wrappers=1:nokey=1",
                video_path
            ]
            result = subprocess.run(cmd, capture_output=True, text=True, check=True)
            return float(result.stdout.strip())
        except Exception as e:
            print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
            return None

    def apply_sound_effects_to_folder(folder_path):
        EX_KEY = userdata.get('ELEVENLABS_API_KEY')

        if not EX_KEY:
            print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
            return []

        try:
            elevenlabs_client = ElevenLabs(api_key=EX_KEY)
        except Exception as e:
            print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
            return []

        if not os.path.exists(folder_path):
            print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
            return []

        script_path = os.path.join(folder_path, "production_script.json")
        scenes = {}
        if os.path.exists(script_path):
            try:
                with open(script_path, 'r', encoding='utf-8') as f:
                    raw_script = json.load(f)
                    if isinstance(raw_script, list):
                        scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                    else:
                        print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
            except json.JSONDecodeError as e:
                print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
            except Exception as e:
                print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

        videos_to_process = [
            f for f in os.listdir(folder_path)
            if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and not f.endswith('_AUDIO_READY.mp4')
        ]
        print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

        temp_audio_dir = "/tmp/elevenlabs_temp_audio"
        os.makedirs(temp_audio_dir, exist_ok=True)
        print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

        processed_video_files = []

        for vid_filename in videos_to_process:
            s_id_match = re.search(r'scene_(\d+)_VIDEO\\.mp4', vid_filename)
            scene_id = int(s_id_match.group(1)) if s_id_match else None

            v_full_path = os.path.join(folder_path, vid_filename)
            v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
            sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
            final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

            if os.path.exists(final_video_with_sfx):
                print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
                processed_video_files.append(final_video_with_sfx)
                continue

            video_duration = get_video_duration(v_full_path)
            if video_duration is None:
                print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
                continue

            scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
            sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
                "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
            )

            print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

            try:
                audio_response = elevenlabs_client.text_to_sound_effects.convert(
                    text=sfx_prompt,
                    duration_seconds=min(video_duration + 1.0, 60.0),
                    prompt_influence=0.6,
                    output_format="mp3_44100_128"
                )

                with open(sfx_audio_file, "wb") as f:
                    for chunk in audio_response:
                        f.write(chunk)
                print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

                print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
                ffmpeg_command = [
                    "ffmpeg", "-y",
                    "-i", v_full_path,
                    "-i", sfx_audio_file,
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-shortest",
                    final_video_with_sfx
                ]

                ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
                print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
                processed_video_files.append(final_video_with_sfx)

            except requests.exceptions.RequestException as e:
                print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
                print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
            except Exception as e:
                print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
            finally:
                if os.path.exists(sfx_audio_file):
                    os.remove(sfx_audio_file)
                    print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

        if os.path.exists(temp_audio_dir) and not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)

        print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
        return processed_video_files

    def generate_scene_music(scene_id, folder_path, music_prompt_text):
        EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
        if not EL_API_KEY:
            print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
            return

        client = ElevenLabs(api_key=EL_API_KEY)

        output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

        print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt_text}'", flush=True)
        try:
            track = client.music.compose(
                prompt=music_prompt_text,
                music_length_ms=7000
            )

            with open(output_file, "wb") as f:
                for chunk in track:
                    f.write(chunk)
            print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
        except Exception as e:
            print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

    def create_final_master_video(project_folder_path):
        if not os.path.exists('/content/drive'):
            print("🚀 Mounting Google Drive...", flush=True)
            drive.mount('/content/drive')

        if not os.path.exists(project_folder_path):
            print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
            return

        print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

        def get_scene_id_from_filename(filename, pattern):
            match = re.search(pattern, filename)
            if match:
                scene_id = int(match.group(1))
                return scene_id
            return 0

        all_files = os.listdir(project_folder_path)

        audio_ready_videos = [
            f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
        ]
        audio_ready_videos.sort(key=lambda x: get_scene_id_from_filename(x, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4'))


        if not audio_ready_videos:
            print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found or valid scene IDs extracted in {project_folder_path}. Aborting final merge.", flush=True)
            return

        final_scene_videos_for_concat = []
        temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
        os.makedirs(temp_audio_mix_dir, exist_ok=True)

        for vid_filename in audio_ready_videos:
            s_id_match = re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4', vid_filename)
            if not s_id_match:
                print(f"⚠️ Could not extract scene ID from {vid_filename}. Skipping this file.", flush=True)
                continue
            scene_id = int(s_id_match.group(1))

            input_video_path = os.path.join(project_folder_path, vid_filename)
            music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
            output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

            if os.path.exists(music_file_path):
                print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

                temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
                try:
                    subprocess.run([
                        "ffmpeg", "-y", "-i", input_video_path,
                        "-map", "0:a:0", temp_ambient_audio
                    ], capture_output=True, text=True, check=True)
                except subprocess.CalledProcessError as e:
                    print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                    shutil.copy(input_video_path, output_final_scene_video_path)
                    final_scene_videos_for_concat.append(output_final_scene_video_path)
                    continue

                mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
                try:
                    subprocess.run([
                        "ffmpeg", "-y",
                        "-i", temp_ambient_audio,
                        "-i", music_file_path,
                        "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                        "-map", "[aout]", mixed_audio_output
                    ], capture_output=True, text=True, check=True)
                except subprocess.CalledProcessError as e:
                    print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                    shutil.copy(input_video_path, output_final_scene_video_path)
                    final_scene_videos_for_concat.append(output_final_scene_video_path)
                    continue

                try:
                    subprocess.run([
                        "ffmpeg", "-y",
                        "-i", input_video_path,
                        "-i", mixed_audio_output,
                        "-map", "0:v:0",
                        "-map", "1:a:0",
                        "-c:v", "copy",
                        "-c:a", "aac",
                        "-b:a", "192k",
                        "-shortest",
                        output_final_scene_video_path
                    ], capture_output=True, text=True, check=True)
                    final_scene_videos_for_concat.append(output_final_scene_video_path)
                    print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
                except subprocess.CalledProcessError as e:
                    print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                    final_scene_videos_for_concat.append(output_final_scene_video_path)
                    shutil.copy(input_video_path, output_final_scene_video_path)
            else:
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

        if not final_scene_videos_for_concat:
            print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
            return

        list_file = os.path.join(project_folder_path, "concat_final.txt")
        with open(list_file, "w") as f:
            for v in sorted(final_scene_videos_for_concat, key=lambda x: get_scene_id_from_filename(os.path.basename(x), r'scene_(\d+)_FINAL\\.mp4')):
                f.write(f"file '{os.path.basename(v)}'\n")

        output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
        output_master_path = os.path.join(project_folder_path, output_master_filename)

        print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

        try:
            subprocess.run(
                f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
                shell=True,
                check=True,
                cwd=project_folder_path,
                capture_output=True
            )
            print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
            print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
        except FileNotFoundError:
            print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
        finally:
            if os.path.exists(temp_audio_mix_dir):
                shutil.rmtree(temp_audio_mix_dir)
                print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)


    # Get dynamic models
    MODEL_BRAIN_SELECTED, MODEL_ASSISTANT_SELECTED = get_best_models(G_KEY)

    print(f"🛠️ MODE: {'[TEST/DRY-RUN]' if config['test_mode'] else '[LIVE PRODUCTION]'}", flush=True)
    print(f"🧠 BRAIN: {MODEL_BRAIN_SELECTED} | ⚡ ASSISTANT: {MODEL_ASSISTANT_SELECTED}", flush=True)

    SAFETY_SETTINGS = [
        {"category": "HARM_CATEGORY_HARASSMENT", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_HATE_SPEECH", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_SEXUALLY_EXPLICIT", "threshold": "BLOCK_NONE"},
        {"category": "HARM_CATEGORY_DANGEROUS_CONTENT", "threshold": "BLOCK_NONE"}
    ]

    # 1. NAME & FOLDER
    print("🧠 Brainstorming Episode & Toy Name...", flush=True)
    raw_n = gemini_call(f"Generate a creative, attractive 3-word name for a colossal, surreal water park attraction, similar to 'Deep Vortex' or 'Vacuum Flush'. Output ONLY the name.", G_KEY, model=MODEL_BRAIN_SELECTED)
    episode_attractive_name = re.sub(r'[\\W_]+', ' ', (raw_n or "Dream_Slide")).strip().replace(' ', '_')
    timestamp = time.strftime('%Y%m%d_%H%M%S')
    project_folder_name = f"{episode_attractive_name}_{timestamp}"

    if not os.path.exists('/content/drive'): drive.mount('/content/drive')
    path = f"/content/drive/MyDrive/{config['root_folder']}/{project_folder_name}"
    if not config['test_mode']: os.makedirs(path, exist_ok=True)
    print(f"🚀 PRODUCTION EPISODE: '{episode_attractive_name.replace('_', ' ')}' saved to: {path}", flush=True)

    # 2. SCRIPTING
    print("📝 Generating script...", flush=True)
    narrative_intents_str = "\n".join([f"- Scene {i+1}: {intent}" for i, intent in enumerate(NARRATIVE_INTENTS)])

    script_prompt = (f"As a Master Storyboard Artist, create a {config['scene_count']}-scene JSON list for the episode titled '{episode_attractive_name.replace('_', ' ')}'.\n"\
                     f"Theme: {config['theme']}\n"\
                     f"The central attraction for this episode is the '{episode_attractive_name.replace('_', ' ')}'.\n"\
                     f"Each scene's description MUST be hyper-realistic and include specific details from '{config['material_water']}', '{config['material_structures']}' and '{config['persona']}'.\n"\
                     f"The 'audio_prompt' for each scene MUST reflect the presence of the female persona and constant background laughter and ambient water park sounds. For scene 9, specifically integrate the retro vintage music from PROJECT_CONFIG['music_prompt_base'] into the 'audio_prompt'.\n"\
                     f"Follow these narrative intents for each scene:\n"\
                     f"{narrative_intents_str}\n"\
                     f"Output JSON list where each object has: 'scene_id', 'title', 'image_logic' (detailed visual description), 'motion_logic' (detailed camera movement), 'audio_prompt' (sound design).")

    raw_j = gemini_call(script_prompt, G_KEY, is_json=True, model=MODEL_BRAIN_SELECTED)
    if not raw_j: return print("❌ Script Failure: Gemini did not return a valid script.", flush=True)

    try:
        cleaned_json_str = re.sub(r'```json\\n?|```', '', raw_j).strip()
        scenes = json.loads(cleaned_json_str)

        if not isinstance(scenes, list):
            print("⚠️ Warning: Script did not return a list. Attempting to convert.", flush=True)
            scenes = [scenes]

        for i, scene in enumerate(scenes):
            if 'scene_id' not in scene or not isinstance(scene['scene_id'], (int, str)):
                scene['scene_id'] = i + 1
            else:
                try:
                    scene['scene_id'] = int(scene['scene_id'])
                except (ValueError, TypeError):
                    scene['scene_id'] = i + 1

    except json.JSONDecodeError as e:
        print(f"❌ JSON Parsing Error: {e}", flush=True)
        print(f"Raw JSON string received: {raw_j}", flush=True)
        return
    except Exception as e:
        print(f"❌ Unexpected Error during JSON parsing: {e}", flush=True)
        return

    if not config['test_mode']:
        script_file_path = os.path.join(path, "production_script.json")
        with open(script_file_path, "w", encoding='utf-8') as f:
            json.dump(scenes, f, indent=4, ensure_ascii=False)
        print(f"✅ Production script saved to: {script_file_path}", flush=True)

    video_paths = []

    # 3. PRODUCTION LOOP
    for scene in scenes:
        s_id = scene.get('scene_id', 'X')
        print(f"\n🎬 Scene {s_id}: {scene.get('title')}", flush=True)

        if not config['test_mode']:
            print(f"🖼️ Image Generation Model: {MODEL_ASSISTANT_SELECTED}", flush=True)
            img_prompt = (
                f"'{scene.get('image_logic')}.' "
                f"Theme: {config['theme']}. "
                f"Lighting: {config['lighting']}. "
                f"Materials: {config['material_water']} with ultra-realistic physics and {config['material_structures']} with hyper-realistic structural integrity. "
                f"Persona: {config['persona']}. "
                f"Aspect Ratio: {config['aspect_ratio']}. "
                f"{config['image_enhancements']}. "
                f"Negative prompt: {config['negative_prompt']}"
            )

            img_res = call_gemini_image(img_prompt, G_KEY, MODEL_ASSISTANT_SELECTED, s_id)

            if img_res:
                candidate = img_res['candidates'][0]
                img_part = next((p for p in candidate['content']['parts'] if 'inlineData' in p), None)
                img_data = base64.b64decode(img_part['inlineData']['data'])
                img_file = f"{path}/scene_{s_id}.png"
                with open(img_file, "wb") as f: f.write(img_data)
                display(Image(filename=img_file, width=400))

                print(f"⬆️ Uploading image for scene {s_id} to ImgBB...", flush=True)
                imgbb_url = "https://api.imgbb.com/1/upload"
                imgbb_files = {'image': open(img_file, 'rb')}
                imgbb_payload = {'key': I_KEY, 'expiration': 3600}
                try:
                    imgbb_res = requests.post(imgbb_url, files=imgbb_files, data=imgbb_payload).json()
                    url_img = imgbb_res["data"]["url"]
                    print(f"✅ ImgBB URL obtained for scene {s_id}.", flush=True)
                except Exception as e:
                    print(f"❌ ImgBB upload failed for scene {s_id}: {e}", flush=True)
                    url_img = None

                if url_img:
                    payload_kling = {
                        "model": M_VID,
                        "prompt": f"{scene.get('motion_logic')}, ultra-realistic physics and material interactions, {config['material_water']} with precise physics, {config['material_structures']} with structural realism, {scene.get('audio_prompt')}, {config['persona']}, {config['style']}",
                        "image_start": url_img, "duration": 7, "quality": "720p", "sound": "on" if config['audio_enabled'] else "off"
                    }
                    kling_headers = {"Authorization": f"Bearer {K_KEY}", "Content-Type": "application/json"}
                    kv_res = requests.post(f"{B_URL}/videos/generations", json=payload_kling, headers=kling_headers).json()
                    task_id = kv_res.get('id') or kv_res.get('data', {}).get('id')

                    if task_id:
                        print(f"🚀 Kling Task {task_id} Created for scene {s_id}. Monitoring...", flush=True)
                        endpoints = [f"{B_URL}/videos/generations/{task_id}", f"{B_URL}/tasks/{task_id}"]
                        idx_endpoint = 0
                        while True:
                            try:
                                mon_res = requests.get(endpoints[idx_endpoint], headers=kling_headers)
                                if mon_res.status_code == 404 and idx_endpoint == 0:
                                    idx_endpoint = 1; continue

                                if mon_res.status_code == 200:
                                    data = mon_res.json()
                                    status = str(data.get("status") or data.get("data", {}).get("status") or "").lower()
                                    print(f"[{status.upper() or 'WAITING'}]", end=" ", flush=True)

                                    if status in ["completed", "succeeded", "finished"]:
                                        video_url = extrair_url_video(data)
                                        if video_url:
                                            v_file = f"{path}/scene_{s_id}_VIDEO.mp4"
                                            print(f"\n📥 Downloading video for scene {s_id} from {video_url}", flush=True)
                                            with requests.get(video_url, stream=True, timeout=120) as r:
                                                r.raise_for_status()
                                                with open(v_file, 'wb') as f: shutil.copyfileobj(r.raw, f)
                                            if os.path.exists(v_file) and os.path.getsize(v_file) > 1000:
                                                video_paths.append(v_file)
                                                print(f"✅ Video for scene {s_id} saved to Drive.", flush=True)
                                                break
                                            else:
                                                print(f"❌ Downloaded video for scene {s_id} is too small or corrupted.", flush=True)
                                                break
                                        else:
                                            print(f"\n❌ Video URL not found in response for scene {s_id}.", flush=True)
                                            break
                                    elif status in ["failed", "error"]:
                                        print(f"\n❌ Kling Error for scene {s_id}: {data}", flush=True); break
                                time.sleep(25)
                            except requests.exceptions.RequestException as e:
                                print(f"\n⚠️ Network/API Error while monitoring scene {s_id}: {e}", flush=True); time.sleep(20)
                            except Exception as e:
                                print(f"\n⚠️ Unexpected Error while monitoring scene {s_id}: {e}", flush=True); time.sleep(20)
                    else:
                        print(f"❌ Failed to get Kling task ID for scene {s_id}: {kv_res}", flush=True)
                else:
                    print(f"❌ Skipping video generation for scene {s_id} due to ImgBB upload failure.", flush=True)
            else:
                print(f"❌ Image generation failed for scene {s_id} after retries.", flush=True)
        else:
            print(f"   🖼️ [PROMPT]: {scene.get('image_logic')}", flush=True)
            print(f"   🎥 [MOVE]: {scene.get('motion_logic')}", flush=True)
            print(f"   🎵 [AUDIO]: {scene.get('audio_prompt')}", flush=True)

    # 4. FINAL STITCH (Audio Processing and Concatenation)
    if not config['test_mode']:
        if config['elevenlabs_sfx_enabled']:
            print(f"\n🔊 Applying Eleven Labs ambient sound effects to videos in {path}...", flush=True)
            audio_ready_video_paths = apply_sound_effects_to_folder(path)
            print(f"✅ Ambient sound effects applied to {len(audio_ready_video_paths)} videos.", flush=True)

            # Generate scene-specific music for scenes 1, 4, and 9
            music_scenes_to_generate = {
                1: config['music_prompt_base'],
                4: config['music_prompt_base'].replace('orchestral piece', 'tropical exotica lounge music, slow tempo, vibraphone and muffled bongos, heavy reverb'),
                9: config['music_prompt_base'].replace('mysterious and awe, slightly eerie', 'majestic, uplifting, expansive orchestral track, concluding a summer adventure, 1950s big band feel')
            }

            for s_id, music_prompt in music_scenes_to_generate.items():
                generate_scene_music(s_id, path, music_prompt)

            create_final_master_video(path)
        else:
            print("\n⚠️ Eleven Labs audio processing is disabled. Skipping audio effects and music generation.", flush=True)

            if video_paths:
                output_filename = f"{project_folder_name}_MASTER.mp4"
                output = os.path.join(path, output_filename)
                list_file = os.path.join(path, "concat.txt")
                with open(list_file, "w") as f:
                    for v in video_paths: f.write(f"file '{os.path.basename(v)}'\n")
                try:
                    print(f"\n✨ Concatenating {len(video_paths)} videos into {output_filename}...", flush=True)
                    subprocess.run(
                        f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output}\"",
                        shell=True,
                        check=True,
                        cwd=path,
                        capture_output=True
                    )
                    print(f"\n🏆 PRODUCTION COMPLETE: {output}", flush=True)
                except subprocess.CalledProcessError as e:
                    print(f"❌ FFmpeg concatenation failed: {e}", flush=True)
                    print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
                except FileNotFoundError:
                    print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
            else:
                print("⚠️ No videos were generated for concatenation.", flush=True)
    elif config['test_mode']:
        print(f"\n✅ EPISODE '{project_folder_name}' READY FOR PRODUCTION (TEST MODE).", flush=True)
    else:
        print(f"\n⚠️ No videos were generated.", flush=True)

if __name__ == "__main__":
    run_factory(PROJECT_CONFIG)
````

### Helper Function: Generate & Merge Eleven Labs TTS (Single Video)

````python
import os
import requests
import subprocess
import time
from google.colab import userdata

# --- [5. ELEVEN LABS AUDIO GENERATION & MERGE FUNCTION] ---

def generate_and_merge_elevenlabs_audio(video_file_path, audio_prompt, output_video_path, voice_id="21m00Tcm4obs8fb3rGRf", elevenlabs_api_key=None, temp_audio_dir="/tmp"):
    """
    Generates audio using Eleven Labs and merges it with a given video file.

    Args:
        video_file_path (str): Path to the input video file (e.g., scene_X_VIDEO.mp4).
        audio_prompt (str): Text to be converted to speech by Eleven Labs.
        output_video_path (str): Path where the new video with merged audio will be saved.
        voice_id (str): The voice ID from Eleven Labs to use for speech generation.
        elevenlabs_api_key (str, optional): Your Eleven Labs API key. If None, it will try to get from userdata.
        temp_audio_dir (str): Directory to save temporary audio files.

    Returns:
        str: Path to the output video with merged audio if successful, None otherwise.
    """
    print(f"🎵 Generating audio for: {video_file_path} with prompt: '{audio_prompt[:50]}...'", flush=True)

    if elevenlabs_api_key is None:
        elevenlabs_api_key = userdata.get('ELEVENLABS_API_KEY')

    if not elevenlabs_api_key:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return None

    headers = {
        "Accept": "audio/mpeg",
        "Content-Type": "application/json",
        "xi-api-key": elevenlabs_api_key
    }

    data = {
        "text": audio_prompt,
        "model_id": "eleven_multilingual_v2",
        "voice_settings": {
            "stability": 0.5,
            "similarity_boost": 0.75
        }
    }

    tts_url = f"https://api.elevenlabs.io/v1/text-to-speech/{voice_id}"
    temp_audio_file = os.path.join(temp_audio_dir, f"temp_audio_{int(time.time())}.mp3")

    try:
        response = requests.post(tts_url, headers=headers, json=data)
        response.raise_for_status() # Raise an exception for HTTP errors (4xx or 5xx)

        with open(temp_audio_file, 'wb') as f:
            f.write(response.content)
        print(f"    ✅ Audio generated and saved to: {temp_audio_file}", flush=True)

        # Now merge audio with video using ffmpeg
        # Use -y to overwrite output files without asking
        # Use -map 0:v:0 to select video stream from input 0
        # Use -map 1:a:0 to select audio stream from input 1
        # -c:v copy to copy video stream directly
        # -c:a aac to re-encode audio to AAC (common format)
        # -shortest to stop when the shortest input stream ends

        ffmpeg_command = [
            "ffmpeg", "-y",
            "-i", video_file_path,  # Input video
            "-i", temp_audio_file,  # Input audio
            "-map", "0:v:0", "-map", "1:a:0",
            "-c:v", "copy",
            "-c:a", "aac",
            "-shortest",
            output_video_path
        ]

        print(f"     merging audio with video using ffmpeg...", flush=True)
        result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
        print(f"    ffmpeg stdout: {result.stdout}", flush=True)
        print(f"    ffmpeg stderr: {result.stderr}", flush=True)

        print(f"    ✅ Audio merged successfully to: {output_video_path}", flush=True)
        return output_video_path

    except requests.exceptions.RequestException as e:
        print(f"❌ Eleven Labs API error: {e}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg error during merge: {e}", flush=True)
        print(f"    FFmpeg stdout: {e.stdout}", flush=True)
        print(f"    FFmpeg stderr: {e.stderr}", flush=True)
    except Exception as e:
        print(f"❌ An unexpected error occurred: {e}", flush=True)
    finally:
        if os.path.exists(temp_audio_file):
            os.remove(temp_audio_file)
            print(f"    Cleaned up temporary audio file: {temp_audio_file}", flush=True)

    return None
````

### Function: Apply Eleven Labs Sound Effects to Folder (Bulk)

### New Modular Function: Applying Eleven Labs Sound Effects to Video Clips

This function (`apply_sound_effects_to_folder`) automates the process of generating ambient sound effects using Eleven Labs and merging them into your individual video clips. It dynamically determines video duration, requests appropriate sound effects, and handles the `ffmpeg` merging process.

**To use this function:**

1.  **Ensure `ELEVENLABS_API_KEY` is set** in your Colab secrets.
2.  **Replace `TARGET_FOLDER`** in the example call below with the full path to the project folder containing your `_VIDEO.mp4` clips.
3.  **Run the cell** to apply sound effects to all un-processed videos in that folder.

````python
import os, requests, subprocess, json, time
from google.colab import drive, userdata

# Helper function to get video duration using ffprobe
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and 'AUDIO_READY' not in f
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    for vid_filename in videos_to_process:
        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            continue

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # 3. Generate Audio via Sound Effects v2
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s)...", flush=True)

        # Enhanced prompt for more specific sound effects
        sfx_prompt = "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."

        url = "https://api.elevenlabs.io/v1/sound-effects"
        headers = {
            "xi-api-key": EX_KEY,
            "Content-Type": "application/json"
        }
        data = {
            "text": sfx_prompt,
            "duration_seconds": min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
            "prompt_influence": 0.6, # Slightly increased for more influence
            "output_format": "mp3_44100_128"
        }

        try:
            res = requests.post(url, json=data, headers=headers)
            res.raise_for_status() # Raise an exception for HTTP errors (4xx or 5xx)

            with open(sfx_audio_file, "wb") as f:
                f.write(res.content)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if res.status_code != 200:
                print(f"    Response content: {res.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}", flush=True)
            print(f"    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if not os.listdir(temp_audio_dir):
        os.rmdir(temp_audio_dir)
        print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
    else:
        print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)

# --- Example Usage (Remember to replace TARGET_FOLDER) ---
# TARGET_FOLDER = "/content/drive/MyDrive/Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise/Your_Toy_Folder"
# apply_sound_effects_to_folder(TARGET_FOLDER)
````

````python
import os, requests, subprocess, json, time
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration using ffprobe
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and 'AUDIO_READY' not in f
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            continue

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...'", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}", flush=True)
            print(f"    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if not os.listdir(temp_audio_dir):
        os.rmdir(temp_audio_dir)
        print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
    else:
        print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
````

````python
import os
from elevenlabs.client import ElevenLabs
from google.colab import userdata

def generate_scene_music(scene_id, folder_path):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    music_prompt = ""
    if scene_id == 1:
        music_prompt = "A grand, awe-inspiring, mysterious orchestral piece, 1950s cinematic feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 4:
        music_prompt = "1950s tropical exotica lounge music, slow tempo, vibraphone and muffled bongos, heavy reverb, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 9:
        music_prompt = "A majestic, uplifting, expansive orchestral track, concluding a summer adventure, 1950s big band feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    else:
        print(f"⚠️ No specific music prompt for scene {scene_id}. Skipping music generation.", flush=True)
        return

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)
````

### New Function: Create Final Master Video with All Audio Layers

This function (`create_final_master_video`) integrates all generated audio layers (ambient sound effects and scene-specific music) with their corresponding video clips and then concatenates them into a single, comprehensive master video.

**Process:**
1.  **Identify Video and Music Files:** It looks for `_VIDEO_AUDIO_READY.mp4` (videos with ambient SFX) and `_music.mp3` files in the specified project folder.
2.  **Merge Video + Ambient SFX + Music:** For scenes that have both an `_AUDIO_READY.mp4` and a `_music.mp3`, it uses `ffmpeg` to mix these two audio tracks and merge them with the video stream, creating `_FINAL.mp4` files.
3.  **Prepare Non-Music Scenes:** For scenes with only `_AUDIO_READY.mp4` (no specific music), it renames them to `_FINAL.mp4`.
4.  **Concatenate Final Videos:** Finally, it concatenates all these `_FINAL.mp4` files, in order, into the ultimate `_MASTER_FINAL_AUDIO.mp4`.

````python
import os
import subprocess
import re
import shutil
from google.colab import drive

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # 1. Collect and prepare all scene video files
    all_files = os.listdir(project_folder_path)
    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=lambda x: int(re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', x).group(1)))

    if not audio_ready_videos:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found in {project_folder_path}. Aborting final merge.", flush=True)
        return

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        s_id_match = re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', vid_filename)
        if not s_id_match: continue
        scene_id = int(s_id_match.group(1))

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                final_scene_videos_for_concat.append(input_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                final_scene_videos_for_concat.append(input_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(input_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: int(re.search(r'scene_(\d+)_FINAL\.mp4', x).group(1))):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)


# --- Example Usage for Music Generation ---
# Define the base project root and target folder
PROJECT_ROOT_CONFIG = "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise"
BASE_PATH = f"/content/drive/MyDrive/{PROJECT_ROOT_CONFIG}"

# The specific project folder you want to save the music files
TARGET_FOLDER_KODACHROME = os.path.join(BASE_PATH, "Kodachrome_Gravity_Glaze_20260416_205630")

# NOTE: Replace with the actual project folder where you want to save the music files
# This should be the folder containing your _VIDEO_AUDIO_READY.mp4 files and _music.mp3 files
TARGET_FOLDER_FINAL_MERGE = TARGET_FOLDER_KODACHROME

print(f"\nAttempting to create final master video for: {TARGET_FOLDER_FINAL_MERGE}\n", flush=True)
create_final_master_video(TARGET_FOLDER_FINAL_MERGE)
````

````python
import os, requests, subprocess, json, time
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration using ffprobe
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and 'AUDIO_READY' not in f
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            continue

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...'", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}", flush=True)
            print(f"    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if not os.listdir(temp_audio_dir):
        os.rmdir(temp_audio_dir)
        print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
    else:
        print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
````

````python
import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return [] # Return an empty list if API key is missing

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return [] # Return an empty list if client initialization fails

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return [] # Return an empty list if folder not found

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and 'AUDIO_READY' not in f
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    processed_video_files = [] # To store paths of successfully processed AUDIO_READY files

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            processed_video_files.append(final_video_with_sfx)
            continue

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
            processed_video_files.append(final_video_with_sfx) # Add to list if successful

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if os.path.exists(temp_audio_dir):
        if not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
        else:
            print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
    return processed_video_files # Return the list of processed files

def generate_scene_music(scene_id, folder_path, music_prompt_text):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
        music_prompt_text (str): The specific text prompt for music generation.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt_text}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt_text,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # 1. Collect and prepare all scene video files
    all_files = os.listdir(project_folder_path)
    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=lambda x: int(re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4', x).group(1)) if re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4', x) else 0)

    if not audio_ready_videos:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found in {project_folder_path}. Aborting final merge.", flush=True)
        return

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        s_id_match = re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4', vid_filename)
        if not s_id_match:
            print(f"⚠️ Could not extract scene ID from {vid_filename}. Skipping this file.", flush=True)
            continue
        scene_id = int(s_id_match.group(1))

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Use the intended output name for consistency
                shutil.copy(input_video_path, output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Use the intended output name for consistency
                shutil.copy(input_video_path, output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: int(re.search(r'scene_(\d+)_FINAL\\.mp4', x).group(1)) if re.search(r'scene_(\d+)_FINAL\\.mp4', x) else 0):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
import os
import subprocess
import re
import shutil
from google.colab import drive

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # 1. Collect and prepare all scene video files
    all_files = os.listdir(project_folder_path)
    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=lambda x: int(re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', x).group(1)))

    if not audio_ready_videos:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found in {project_folder_path}. Aborting final merge.", flush=True)
        return

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        s_id_match = re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', vid_filename)
        if not s_id_match: continue
        scene_id = int(s_id_match.group(1))

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                final_scene_videos_for_concat.append(input_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                final_scene_videos_for_concat.append(input_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(input_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: int(re.search(r'scene_(\d+)_FINAL\.mp4', x).group(1))):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return [] # Return an empty list if API key is missing

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return [] # Return an empty list if client initialization fails

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return [] # Return an empty list if folder not found

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and not f.endswith('_AUDIO_READY.mp4')
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    processed_video_files = [] # To store paths of successfully processed AUDIO_READY files

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        # Debug prints added here
        print(f"  DEBUG: Checking for {final_video_with_sfx}", flush=True)
        print(f"  DEBUG: os.path.exists({final_video_with_sfx}) returned {os.path.exists(final_video_with_sfx)}", flush=True)

        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            processed_video_files.append(final_video_with_sfx)
            continue # Skip to next video if AUDIO_READY already exists

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
            processed_video_files.append(final_video_with_sfx) # Add to list if successful

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if os.path.exists(temp_audio_dir):
        if not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
        else:
            print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
    return processed_video_files # Return the list of processed files

def generate_scene_music(scene_id, folder_path):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    music_prompt = ""
    if scene_id == 1:
        music_prompt = "A grand, awe-inspiring, mysterious orchestral piece, 1950s cinematic feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 4:
        music_prompt = "1950s tropical exotica lounge music, slow tempo, vibraphone and muffled bongos, heavy reverb, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 9:
        music_prompt = "A majestic, uplifting, expansive orchestral track, concluding a summer adventure, 1950s big band feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    else:
        print(f"⚠️ No specific music prompt for scene {scene_id}. Skipping music generation.", flush=True)
        return

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # 1. Collect and prepare all scene video files
    all_files = os.listdir(project_folder_path)
    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX

    # Robust sorting: default to 0 if regex doesn't match
    def get_sort_key(filename, pattern):
        match = re.search(pattern, filename)
        return int(match.group(1)) if match else 0

    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=lambda x: get_sort_key(x, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')) # Use robust helper

    if not audio_ready_videos:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found in {project_folder_path}. Aborting final merge.", flush=True)
        return

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        scene_id = get_sort_key(vid_filename, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')
        if scene_id == 0:
            print(f"⚠️ Could not extract valid scene ID from {vid_filename}. Skipping this file for audio mixing.", flush=True)
            continue

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: get_sort_key(os.path.basename(x), r'scene_(\d+)_FINAL\\.mp4')):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return [] # Return an empty list if API key is missing

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return [] # Return an empty list if client initialization fails

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return [] # Return an empty list if folder not found

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and not f.endswith('_AUDIO_READY.mp4')
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    processed_video_files = [] # To store paths of successfully processed AUDIO_READY files

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        # Debug prints added here
        print(f"  DEBUG: Checking for {final_video_with_sfx}", flush=True)
        print(f"  DEBUG: os.path.exists({final_video_with_sfx}) returned {os.path.exists(final_video_with_sfx)}", flush=True)

        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            processed_video_files.append(final_video_with_sfx)
            continue # Skip to next video if AUDIO_READY already exists

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
            processed_video_files.append(final_video_with_sfx) # Add to list if successful

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if os.path.exists(temp_audio_dir):
        if not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
        else:
            print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
    return processed_video_files # Return the list of processed files

def generate_scene_music(scene_id, folder_path, music_prompt_text):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
        music_prompt_text (str): The specific text prompt for music generation.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt_text}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt_text,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # Helper to get scene ID from filename, handles missing matches
    def get_scene_id_from_filename(filename, pattern):
        print(f"  DEBUG: get_scene_id_from_filename received filename: '{filename}', pattern: '{pattern}'", flush=True)
        # Corrected regex to escape the dot with a single backslash
        match = re.search(pattern, filename)
        if match:
            scene_id = int(match.group(1))
            print(f"  DEBUG: Matched scene ID: {scene_id}", flush=True)
            return scene_id
        else:
            print(f"  DEBUG: No match for '{filename}' with pattern '{pattern}'", flush=True)
            return 0 # Indicate no valid ID found

    all_files = os.listdir(project_folder_path)

    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    # Filter and sort robustly
    audio_ready_videos = []
    for f in all_files:
        if f.endswith('_VIDEO_AUDIO_READY.mp4'):
            # Corrected regex pattern here
            scene_id = get_scene_id_from_filename(f, r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4')
            if scene_id != 0:
                audio_ready_videos.append((scene_id, f))
            else:
                print(f"Warning: Skipping '{f}' for audio mixing and final merge due to invalid scene ID extraction.", flush=True)

    audio_ready_videos.sort(key=lambda x: x[0]) # Sort by scene_id
    audio_ready_filenames = [f[1] for f in audio_ready_videos]

    if not audio_ready_filenames:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found or valid scene IDs extracted in {project_folder_path}. Aborting final merge.", flush=True)
        return

    print(f"DEBUG: Found and sorted {len(audio_ready_filenames)} _AUDIO_READY.mp4 files:", flush=True)
    for f in audio_ready_filenames:
        print(f"  - {f}", flush=True)

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_filenames:
        scene_id = get_scene_id_from_filename(vid_filename, r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4')
        if scene_id == 0: # This should ideally not happen after initial filtering/sorting
            print(f"Warning: Encountered file '{vid_filename}' with invalid scene ID after sorting. Skipping for final merge.", flush=True)
            continue

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: get_scene_id_from_filename(os.path.basename(x), r'scene_(\d+)_FINAL\.mp4')):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
# Re-running function definitions to ensure the latest versions with bug fixes are active
import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and 'AUDIO_READY' not in f
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            continue

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}", flush=True)
            print(f"    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if not os.listdir(temp_audio_dir):
        os.rmdir(temp_audio_dir)
        print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
    else:
        print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)

def generate_scene_music(scene_id, folder_path):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    music_prompt = ""
    if scene_id == 1:
        music_prompt = "A grand, awe-inspiring, mysterious orchestral piece, 1950s cinematic feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 4:
        music_prompt = "1950s tropical exotica lounge music, slow tempo, vibraphone and muffled bongos, heavy reverb, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 9:
        music_prompt = "A majestic, uplifting, expansive orchestral track, concluding a summer adventure, 1950s big band feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    else:
        print(f"⚠️ No specific music prompt for scene {scene_id}. Skipping music generation.", flush=True)
        return

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # 1. Collect and prepare all scene video files
    all_files = os.listdir(project_folder_path)
    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=lambda x: int(re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', x).group(1)) if re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', x) else 0)

    if not audio_ready_videos:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found in {project_folder_path}. Aborting final merge.", flush=True)
        return

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        s_id_match = re.search(r'scene_(\d+)_VIDEO_AUDIO_READY\.mp4', vid_filename)
        if not s_id_match:
            print(f"⚠️ Could not extract scene ID from {vid_filename}. Skipping this file.", flush=True)
            continue
        scene_id = int(s_id_match.group(1))

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Use the intended output name for consistency
                shutil.copy(input_video_path, output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Use the intended output name for consistency
                shutil.copy(input_video_path, output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: int(re.search(r'scene_(\d+)_FINAL\.mp4', x).group(1)) if re.search(r'scene_(\d+)_FINAL\.mp4', x) else 0):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return [] # Return an empty list if API key is missing

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return [] # Return an empty list if client initialization fails

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return [] # Return an empty list if folder not found

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and not f.endswith('_AUDIO_READY.mp4')
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    processed_video_files = [] # To store paths of successfully processed AUDIO_READY files

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        # Debug prints added here
        print(f"  DEBUG: Checking for {final_video_with_sfx}", flush=True)
        print(f"  DEBUG: os.path.exists({final_video_with_sfx}) returned {os.path.exists(final_video_with_sfx)}", flush=True)

        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            processed_video_files.append(final_video_with_sfx)
            continue # Skip to next video if AUDIO_READY already exists

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
            processed_video_files.append(final_video_with_sfx) # Add to list if successful

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if os.path.exists(temp_audio_dir):
        if not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
        else:
            print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
    return processed_video_files # Return the list of processed files

def generate_scene_music(scene_id, folder_path):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    music_prompt = ""
    if scene_id == 1:
        music_prompt = "A grand, awe-inspiring, mysterious orchestral piece, 1950s cinematic feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 4:
        music_prompt = "1950s tropical exotica lounge music, slow tempo, vibraphone and muffled bongos, heavy reverb, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    elif scene_id == 9:
        music_prompt = "A majestic, uplifting, expansive orchestral track, concluding a summer adventure, 1950s big band feel, sounding like it's playing from a distant old speaker at a summer resort, nostalgic and slightly eerie."
    else:
        print(f"⚠️ No specific music prompt for scene {scene_id}. Skipping music generation.", flush=True)
        return

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # Helper to get scene ID from filename, handles missing matches
    def get_scene_id_from_filename(filename, pattern):
        print(f"  DEBUG: get_scene_id_from_filename received filename: '{filename}', pattern: '{pattern}'", flush=True)
        # Corrected regex to escape the dot with a single backslash
        match = re.search(pattern, filename)
        if match:
            scene_id = int(match.group(1))
            print(f"  DEBUG: Matched scene ID: {scene_id}", flush=True)
            return scene_id
        else:
            print(f"  DEBUG: No match for '{filename}' with pattern '{pattern}'", flush=True)
            return 0 # Indicate no valid ID found

    all_files = os.listdir(project_folder_path)

    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    # Filter and sort robustly
    audio_ready_videos = []
    for f in all_files:
        if f.endswith('_VIDEO_AUDIO_READY.mp4'):
            # Corrected regex pattern here
            scene_id = get_scene_id_from_filename(f, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')
            if scene_id != 0:
                audio_ready_videos.append((scene_id, f))
            else:
                print(f"Warning: Skipping '{f}' for audio mixing and final merge due to invalid scene ID extraction.", flush=True)

    audio_ready_videos.sort(key=lambda x: x[0]) # Sort by scene_id
    audio_ready_filenames = [f[1] for f in audio_ready_videos]

    if not audio_ready_filenames:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found or valid scene IDs extracted in {project_folder_path}. Aborting final merge.", flush=True)
        return

    print(f"DEBUG: Found and sorted {len(audio_ready_filenames)} _AUDIO_READY.mp4 files:", flush=True)
    for f in audio_ready_filenames:
        print(f"  - {f}", flush=True)

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_filenames:
        scene_id = get_scene_id_from_filename(vid_filename, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')
        if scene_id == 0: # This should ideally not happen after initial filtering/sorting
            print(f"Warning: Encountered file '{vid_filename}' with invalid scene ID after sorting. Skipping for final merge.", flush=True)
            continue

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: get_scene_id_from_filename(os.path.basename(x), r'scene_(\d+)_FINAL\\.mp4')):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

````python
import os
import re
import subprocess
import shutil
from google.colab import drive, userdata

def create_final_master_video_v2(project_folder_path):
    """
    Enhanced version of the merge function with robust regex for scene IDs.
    """
    if not os.path.exists('/content/drive'):
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Folder not found: {project_folder_path}")
        return

    print(f"🎬 Processing Master for: {os.path.basename(project_folder_path)}")

    all_files = os.listdir(project_folder_path)

    # Robust sorting using a flexible regex to catch scene numbers
    def get_scene_num(filename):
        match = re.search(r'scene_(\d+)', filename)
        return int(match.group(1)) if match else 999

    audio_ready_videos = sorted([
        f for f in all_files if f.endswith('_VIDEO_AUDIO_READY.mp4')
    ], key=get_scene_num)

    if not audio_ready_videos:
        print(f"⚠️ No AUDIO_READY clips found in {os.path.basename(project_folder_path)}.")
        return

    final_scene_videos = []
    temp_mix_dir = os.path.join(project_folder_path, "_temp_final_mix")
    os.makedirs(temp_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_videos:
        scene_id = get_scene_num(vid_filename)
        input_video = os.path.join(project_folder_path, vid_filename)
        music_file = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_scene_final = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file):
            print(f"🎵 Merging Music -> Scene {scene_id}")
            # Extract ambient, mix with music, then merge back to video
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video, "-i", music_file,
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first[a]",
                    "-map", "0:v:0", "-map", "[a]", "-c:v", "copy", "-c:a", "aac", "-shortest",
                    output_scene_final
                ], capture_output=True, check=True)
                final_scene_videos.append(output_scene_final)
            except:
                print(f"⚠️ Audio mix failed for scene {scene_id}, using original.")
                shutil.copy(input_video, output_scene_final)
                final_scene_videos.append(output_scene_final)
        else:
            shutil.copy(input_video, output_scene_final)
            final_scene_videos.append(output_scene_final)

    # Final Concatenation
    list_file = os.path.join(project_folder_path, "concat_final_v2.txt")
    with open(list_file, "w") as f:
        for v in sorted(final_scene_videos, key=get_scene_num):
            f.write(f"file '{os.path.basename(v)}'\n")

    master_name = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_HQ.mp4"
    master_path = os.path.join(project_folder_path, master_name)

    print(f"✨ Concatenating {len(final_scene_videos)} scenes into MASTER...")
    try:
        subprocess.run([
            "ffmpeg", "-y", "-f", "concat", "-safe", "0", "-i", list_file,
            "-c:v", "libx264", "-preset", "fast", "-crf", "23", "-c:a", "aac", master_path
        ], capture_output=True, check=True, cwd=project_folder_path)
        print(f"🏆 MASTER CREATED: {master_name}")
    except Exception as e:
        print(f"❌ Final concat failed: {e}")
    finally:
        if os.path.exists(temp_mix_dir): shutil.rmtree(temp_mix_dir)
````

````python
# Re-running function definitions to ensure the latest versions are active
# This cell re-defines get_video_duration, apply_sound_effects_to_folder, generate_scene_music, and create_final_master_video

import sys
!{sys.executable} -m pip install elevenlabs

import os, requests, subprocess, json, time, re, shutil
from google.colab import drive, userdata
from elevenlabs.client import ElevenLabs

# Helper function to get video duration
def get_video_duration(video_path):
    try:
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format=duration",
            "-of", "default=noprint_wrappers=1:nokey=1",
            video_path
        ]
        result = subprocess.run(cmd, capture_output=True, text=True, check=True)
        return float(result.stdout.strip())
    except Exception as e:
        print(f"❌ Error getting video duration for {video_path}: {e}", flush=True)
        return None

def apply_sound_effects_to_folder(folder_path):
    """
    Generates sound effects for all individual '_VIDEO.mp4' files in a given folder
    using Eleven Labs and merges them, creating '_AUDIO_READY.mp4' files.

    Args:
        folder_path (str): The path to the folder containing the video files.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets. Please add it to Colab secrets.", flush=True)
        return [] # Return an empty list if API key is missing

    # Initialize ElevenLabs client using the SDK
    try:
        elevenlabs_client = ElevenLabs(api_key=EX_KEY)
    except Exception as e:
        print(f"❌ Error initializing ElevenLabs client: {e}", flush=True)
        return [] # Return an empty list if client initialization fails

    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(folder_path):
        print(f"❌ Error: Target folder not found: {folder_path}", flush=True)
        return [] # Return an empty list if folder not found

    script_path = os.path.join(folder_path, "production_script.json")
    scenes = {}
    if os.path.exists(script_path):
        try:
            with open(script_path, 'r', encoding='utf-8') as f:
                raw_script = json.load(f)
                # Ensure raw_script is a list and then convert to dict for easy lookup
                if isinstance(raw_script, list):
                    scenes = {scene['scene_id']: scene for scene in raw_script if 'scene_id' in scene}
                else:
                    print(f"⚠️ production_script.json is not a list. Cannot extract scene-specific audio prompts.", flush=True)
        except json.JSONDecodeError as e:
            print(f"❌ Error decoding production_script.json: {e}. Using generic prompt.", flush=True)
        except Exception as e:
            print(f"❌ Error reading production_script.json: {e}. Using generic prompt.", flush=True)

    # Filter for _VIDEO.mp4 files that are NOT already AUDIO_READY, and exclude MASTER files
    videos_to_process = [
        f for f in os.listdir(folder_path)
        if f.endswith('_VIDEO.mp4') and 'MASTER' not in f and not f.endswith('_AUDIO_READY.mp4')
    ]
    print(f"📂 Found {len(videos_to_process)} muted clips to process for sound effects.", flush=True)

    # Create a temporary directory for audio files
    temp_audio_dir = "/tmp/elevenlabs_temp_audio"
    os.makedirs(temp_audio_dir, exist_ok=True)
    print(f"Created temporary audio directory: {temp_audio_dir}", flush=True)

    processed_video_files = [] # To store paths of successfully processed AUDIO_READY files

    for vid_filename in videos_to_process:
        s_id_match = re.search(r'scene_(\d+)_VIDEO\\.mp4', vid_filename)
        scene_id = int(s_id_match.group(1)) if s_id_match else None

        v_full_path = os.path.join(folder_path, vid_filename)
        v_name_base = vid_filename.rsplit('.', 1)[0] # e.g., "scene_1_VIDEO"
        sfx_audio_file = os.path.join(temp_audio_dir, f"{v_name_base}_sfx_{int(time.time())}.mp3")
        final_video_with_sfx = os.path.join(folder_path, f"{v_name_base}_AUDIO_READY.mp4")

        # Double-check if the AUDIO_READY file already exists to prevent reprocessing
        # Debug prints added here
        print(f"  DEBUG: Checking for {final_video_with_sfx}", flush=True)
        print(f"  DEBUG: os.path.exists({final_video_with_sfx}) returned {os.path.exists(final_video_with_sfx)}", flush=True)

        if os.path.exists(final_video_with_sfx):
            print(f"⏩ Skipping '{vid_filename}': AUDIO_READY file already exists.", flush=True)
            processed_video_files.append(final_video_with_sfx)
            continue # Skip to next video if AUDIO_READY already exists

        video_duration = get_video_duration(v_full_path)
        if video_duration is None:
            print(f"⚠️ Could not determine duration for '{vid_filename}'. Skipping audio generation for this video.", flush=True)
            continue

        # Determine sfx_prompt: use scene-specific if available, else fallback to generic
        scene_specific_audio_prompt = scenes.get(scene_id, {}).get('audio_prompt')
        sfx_prompt = scene_specific_audio_prompt if scene_specific_audio_prompt else (
            "1950s water park resort ambience, heavy viscous liquid splashes, kids laughing, excited screams, summer joy, vintage amusement park background noise, water slides, gushing water."
        )

        # 3. Generate Audio via Sound Effects v2 using the SDK
        print(f"\n🔊 Generating sound effects for: '{vid_filename}' (duration: {video_duration:.2f}s) with prompt: '{sfx_prompt[:70]}...", flush=True)

        try:
            audio_response = elevenlabs_client.text_to_sound_effects.convert(
                text=sfx_prompt,
                duration_seconds=min(video_duration + 1.0, 60.0), # Generate slightly longer than video, up to 60s max for API limit
                prompt_influence=0.6,
                output_format="mp3_44100_128"
            )

            with open(sfx_audio_file, "wb") as f:
                for chunk in audio_response:
                    f.write(chunk)
            print(f"    ✅ Sound effect audio generated and saved to: {sfx_audio_file}", flush=True)

            # 4. Merge with FFmpeg
            print(f"    🎬 Merging sound effects into '{vid_filename}'...", flush=True)
            ffmpeg_command = [
                "ffmpeg", "-y",
                "-i", v_full_path,  # Input video
                "-i", sfx_audio_file,  # Input audio
                "-map", "0:v:0", # Select video stream from first input
                "-map", "1:a:0", # Select audio stream from second input
                "-c:v", "copy", # Copy video stream without re-encoding
                "-c:a", "aac", # Re-encode audio to AAC format
                "-shortest", # Ensure output duration is shortest of input streams
                final_video_with_sfx
            ]

            ffmpeg_result = subprocess.run(ffmpeg_command, capture_output=True, text=True, check=True)
            print(f"    ✅ Successfully merged sound effects to: '{final_video_with_sfx}'", flush=True)
            processed_video_files.append(final_video_with_sfx) # Add to list if successful

        except requests.exceptions.RequestException as e:
            print(f"❌ ElevenLabs API error for '{vid_filename}': {e}", flush=True)
            if 'audio_response' in locals() and hasattr(audio_response, 'text'):
                print(f"    Response content: {audio_response.text}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error during merge for '{vid_filename}': {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ An unexpected error occurred for '{vid_filename}': {e}", flush=True)
        finally:
            # Clean up temporary audio file
            if os.path.exists(sfx_audio_file):
                os.remove(sfx_audio_file)
                print(f"    Cleaned up temporary audio file: {sfx_audio_file}", flush=True)

    # Clean up the temporary audio directory if it's empty
    if os.path.exists(temp_audio_dir):
        if not os.listdir(temp_audio_dir):
            os.rmdir(temp_audio_dir)
            print(f"Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
        else:
            print(f"Temporary audio directory {temp_audio_dir} not empty, skipping cleanup.", flush=True)

    print(f"\n✨ All processable video clips in '{folder_path}' have been processed for sound effects.", flush=True)
    return processed_video_files # Return the list of processed files

def generate_scene_music(scene_id, folder_path, music_prompt_text):
    """
    Generates a specific music track for a given scene ID using ElevenLabs
    and saves it to the specified folder.

    Args:
        scene_id (int): The ID of the scene (1, 4, or 9).
        folder_path (str): The path to the folder where the music should be saved.
        music_prompt_text (str): The specific text prompt for music generation.
    """
    EL_API_KEY = userdata.get("ELEVENLABS_API_KEY")
    if not EL_API_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return

    client = ElevenLabs(api_key=EL_API_KEY)

    output_file = os.path.join(folder_path, f"scene_{scene_id}_music.mp3")

    print(f"🎵 Composing music for scene {scene_id} with prompt: '{music_prompt_text}'", flush=True)
    try:
        # Music length should ideally match the video duration, here assuming 7 seconds
        track = client.music.compose(
            prompt=music_prompt_text,
            music_length_ms=7000 # 7 seconds as per current video clips
        )

        with open(output_file, "wb") as f:
            for chunk in track:
                f.write(chunk)
        print(f"✅ Music for scene {scene_id} saved to: {output_file}", flush=True)
    except Exception as e:
        print(f"❌ Failed to generate music for scene {scene_id}: {e}", flush=True)

def create_final_master_video(project_folder_path):
    """
    Merges individual scene videos (with ambient SFX) and scene-specific music tracks,
    then concatenates them into a final master video.

    Args:
        project_folder_path (str): The absolute path to the project folder.
    """
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')

    if not os.path.exists(project_folder_path):
        print(f"❌ Error: Project directory not found at: {project_folder_path}", flush=True)
        return

    print(f"🎬 Initiating final master video creation for project: {os.path.basename(project_folder_path)}", flush=True)

    # Helper to get scene ID from filename, handles missing matches
    def get_scene_id_from_filename(filename, pattern):
        print(f"  DEBUG: get_scene_id_from_filename received filename: '{filename}', pattern: '{pattern}'", flush=True)
        # Corrected regex to escape the dot with a single backslash
        match = re.search(pattern, filename)
        if match:
            scene_id = int(match.group(1))
            print(f"  DEBUG: Matched scene ID: {scene_id}", flush=True)
            return scene_id
        else:
            print(f"  DEBUG: No match for '{filename}' with pattern '{pattern}'", flush=True)
            return 0 # Indicate no valid ID found

    all_files = os.listdir(project_folder_path)

    # We expect _AUDIO_READY.mp4 files, which already contain ambient SFX
    # Filter and sort robustly
    audio_ready_videos = []
    for f in all_files:
        if f.endswith('_VIDEO_AUDIO_READY.mp4'):
            # Corrected regex pattern here
            scene_id = get_scene_id_from_filename(f, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')
            if scene_id != 0:
                audio_ready_videos.append((scene_id, f))
            else:
                print(f"Warning: Skipping '{f}' for audio mixing and final merge due to invalid scene ID extraction.", flush=True)

    audio_ready_videos.sort(key=lambda x: x[0]) # Sort by scene_id
    audio_ready_filenames = [f[1] for f in audio_ready_videos]

    if not audio_ready_filenames:
        print(f"⚠️ No _VIDEO_AUDIO_READY.mp4 files found or valid scene IDs extracted in {project_folder_path}. Aborting final merge.", flush=True)
        return

    print(f"DEBUG: Found and sorted {len(audio_ready_filenames)} _AUDIO_READY.mp4 files:", flush=True)
    for f in audio_ready_filenames:
        print(f"  - {f}", flush=True)

    final_scene_videos_for_concat = []
    temp_audio_mix_dir = os.path.join(project_folder_path, "_temp_audio_mix")
    os.makedirs(temp_audio_mix_dir, exist_ok=True)

    for vid_filename in audio_ready_filenames:
        scene_id = get_scene_id_from_filename(vid_filename, r'scene_(\d+)_VIDEO_AUDIO_READY\\.mp4')
        if scene_id == 0: # This should ideally not happen after initial filtering/sorting
            print(f"Warning: Encountered file '{vid_filename}' with invalid scene ID after sorting. Skipping for final merge.", flush=True)
            continue

        input_video_path = os.path.join(project_folder_path, vid_filename)
        music_file_path = os.path.join(project_folder_path, f"scene_{scene_id}_music.mp3")
        output_final_scene_video_path = os.path.join(project_folder_path, f"scene_{scene_id}_FINAL.mp4")

        if os.path.exists(music_file_path):
            print(f"🎵 Merging music for scene {scene_id} with video and ambient SFX...", flush=True)

            # Mix the existing audio from _AUDIO_READY.mp4 with the new music.mp3
            # First, extract audio from input_video_path (which has ambient SFX)
            temp_ambient_audio = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_ambient.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y", "-i", input_video_path,
                    "-map", "0:a:0", temp_ambient_audio
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error extracting ambient audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio extraction fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Mix ambient audio with music audio
            mixed_audio_output = os.path.join(temp_audio_mix_dir, f"scene_{scene_id}_mixed_audio.mp3")
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", temp_ambient_audio, # Ambient SFX
                    "-i", music_file_path,    # Scene Music
                    "-filter_complex", "[0:a][1:a]amix=inputs=2:duration=first:dropout_transition=2[aout]",
                    "-map", "[aout]", mixed_audio_output
                ], capture_output=True, text=True, check=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error mixing audio for scene {scene_id}: {e}", flush=True)
                # Fallback: use input_video_path directly if audio mixing fails
                shutil.copy(input_video_path, output_final_scene_video_path)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                continue

            # Merge video with the new mixed audio
            try:
                subprocess.run([
                    "ffmpeg", "-y",
                    "-i", input_video_path, # Video stream from this
                    "-i", mixed_audio_output, # Mixed audio stream from this
                    "-map", "0:v:0",
                    "-map", "1:a:0",
                    "-c:v", "copy",
                    "-c:a", "aac",
                    "-b:a", "192k", # High quality audio bitrate
                    "-shortest", # Ensure output duration is shortest of input streams
                    output_final_scene_video_path
                ], capture_output=True, text=True, check=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path)
                print(f"    ✅ Scene {scene_id} merged with ambient SFX and music.", flush=True)
            except subprocess.CalledProcessError as e:
                print(f"❌ Error merging video with mixed audio for scene {scene_id}: {e}", flush=True)
                final_scene_videos_for_concat.append(output_final_scene_video_path) # Fallback if merge fails, use the video with only ambient sfx.
                shutil.copy(input_video_path, output_final_scene_video_path)
        else:
            # If no specific music, just use the _AUDIO_READY.mp4 as the final scene video
            shutil.copy(input_video_path, output_final_scene_video_path)
            final_scene_videos_for_concat.append(output_final_scene_video_path)
            print(f"    ✅ Scene {scene_id} (ambient SFX only) prepared for final merge.", flush=True)

    if not final_scene_videos_for_concat:
        print("⚠️ No final scene videos to concatenate. Aborting.", flush=True)
        return

    # 2. Concatenate all final scene videos into the ultimate master video
    list_file = os.path.join(project_folder_path, "concat_final.txt")
    with open(list_file, "w") as f:
        # Write relative paths for ffmpeg to work with cwd correctly
        for v in sorted(final_scene_videos_for_concat, key=lambda x: get_scene_id_from_filename(os.path.basename(x), r'scene_(\d+)_FINAL\\.mp4')):
            f.write(f"file '{os.path.basename(v)}'\n")

    output_master_filename = f"{os.path.basename(project_folder_path)}_MASTER_FINAL_AUDIO.mp4"
    output_master_path = os.path.join(project_folder_path, output_master_filename)

    print(f"\n✨ Concatenating {len(final_scene_videos_for_concat)} final scene videos into {output_master_filename}...", flush=True)

    try:
        subprocess.run(
            f"ffmpeg -y -f concat -safe 0 -i \"{list_file}\" -c:v libx264 -preset fast -crf 23 -c:a aac \"{output_master_path}\"",
            shell=True,
            check=True,
            cwd=project_folder_path,
            capture_output=True
        )
        print(f"\n🏆 FINAL MASTER VIDEO CREATED: {output_master_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg final concatenation failed: {e}", flush=True)
        print(f"FFmpeg stdout: {e.stdout.decode()}\nFFmpeg stderr: {e.stderr.decode()}", flush=True)
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please ensure it's installed or available in PATH.", flush=True)
    finally:
        # Clean up temporary audio mix directory
        if os.path.exists(temp_audio_mix_dir):
            shutil.rmtree(temp_audio_mix_dir)
            print(f"✅ Cleaned up temporary audio mix directory: {temp_audio_mix_dir}", flush=True)
````

### Function: Generate & Merge TTS Audio for Master Video

### New Modular Function: Generate and Merge TTS Audio for Master Video

This function (`generate_and_merge_tts_for_folder`) automates the process of generating **Text-to-Speech (TTS)** audio from the `audio_prompt` in your `production_script.json` for each scene, concatenates them, and then merges this combined audio with the final `_MASTER.mp4` video in a specified project folder. This is different from the `apply_sound_effects_to_folder` function, which generates ambient sound effects for individual clips.

**To use this function:**

1.  **Ensure `ELEVENLABS_API_KEY` is set** in your Colab secrets.
2.  **Ensure `ELEVENLABS_VOICE_ID` is set** in your Colab secrets (or use the default in the function).
3.  **Ensure `production_script.json` and `_MASTER.mp4` exist** in the `target_folder_path`.
4.  **Replace `TARGET_FOLDER`** in the example call below with the full path to your project folder.

````python
import os, requests, subprocess, json, time
from google.colab import drive, userdata

def generate_and_merge_tts_for_folder(target_folder_path, scene_audio_duration_sec=7):
    """
    Generates TTS audio for each scene from production_script.json, concatenates them,
    and merges the combined audio with the _MASTER.mp4 in the specified folder.

    Args:
        target_folder_path (str): Path to the project folder containing production_script.json and _MASTER.mp4.
        scene_audio_duration_sec (int): Desired duration for each scene's audio in seconds.
    """
    EX_KEY = userdata.get('ELEVENLABS_API_KEY')
    VOICE_ID = userdata.get('ELEVENLABS_VOICE_ID', '21m00Tcm4obs8fb3rGRf') # Default voice ID

    if not EX_KEY:
        print("❌ Error: ELEVENLABS_API_KEY not found in secrets.", flush=True)
        return
    if not os.path.exists('/content/drive'):
        print("🚀 Mounting Google Drive...", flush=True)
        drive.mount('/content/drive')
    if not os.path.exists(target_folder_path):
        print(f"❌ Error: Target folder not found: {target_folder_path}", flush=True)
        return

    script_path = os.path.join(target_folder_path, "production_script.json")
    master_video_path = os.path.join(target_folder_path, os.path.basename(target_folder_path) + "_MASTER.mp4")
    output_video_with_tts_path = os.path.join(target_folder_path, os.path.basename(target_folder_path) + "_MASTER_WITH_TTS.mp4")

    if not os.path.exists(script_path):
        print(f"❌ Error: production_script.json not found in {target_folder_path}.", flush=True)
        return
    if not os.path.exists(master_video_path):
        print(f"❌ Error: _MASTER.mp4 not found in {target_folder_path}. Make sure it's already concatenated.", flush=True)
        return

    print(f"🎥 Processing folder: {target_folder_path}", flush=True)

    # 1. Load the script
    with open(script_path, 'r', encoding='utf-8') as f:
        scenes = json.load(f)
    print(f"✅ Loaded {len(scenes)} scenes from production_script.json.", flush=True)

    temp_audio_dir = os.path.join(target_folder_path, "_temp_tts_audios")
    os.makedirs(temp_audio_dir, exist_ok=True)
    individual_audio_files = []

    # 2. Generate and trim/pad individual TTS audio files
    for scene in scenes:
        s_id = scene.get('scene_id', 'X')
        audio_prompt = scene.get('audio_prompt', f"No audio prompt for scene {s_id}.")
        temp_raw_audio_file = os.path.join(temp_audio_dir, f"scene_{s_id}_raw.mp3")
        temp_processed_audio_file = os.path.join(temp_audio_dir, f"scene_{s_id}_processed.mp3")

        print(f"\n🔊 Generating TTS for Scene {s_id}: '{audio_prompt[:60]}...'", flush=True)

        tts_url = f"https://api.elevenlabs.io/v1/text-to-speech/{VOICE_ID}"
        headers = {
            "Accept": "audio/mpeg",
            "Content-Type": "application/json",
            "xi-api-key": EX_KEY
        }
        data = {
            "text": audio_prompt,
            "model_id": "eleven_multilingual_v2",
            "voice_settings": {
                "stability": 0.5,
                "similarity_boost": 0.75
            }
        }

        try:
            response = requests.post(tts_url, headers=headers, json=data)
            response.raise_for_status()
            with open(temp_raw_audio_file, 'wb') as f: f.write(response.content)
            print(f"    ✅ Raw audio generated: {temp_raw_audio_file}", flush=True)

            # Trim/pad audio to exact scene_audio_duration_sec
            print(f"    ✂️ Trimming/Padding audio to {scene_audio_duration_sec}s...", flush=True)
            ffmpeg_trim_cmd = [
                "ffmpeg", "-y",
                "-i", temp_raw_audio_file,
                "-af", f"apad=whole_len={scene_audio_duration_sec}:pad_dur={scene_audio_duration_sec},atrim=duration={scene_audio_duration_sec}",
                temp_processed_audio_file
            ]
            subprocess.run(ffmpeg_trim_cmd, capture_output=True, text=True, check=True)
            individual_audio_files.append(temp_processed_audio_file)
            print(f"    ✅ Audio processed to {scene_audio_duration_sec}s: {temp_processed_audio_file}", flush=True)

        except requests.exceptions.RequestException as e:
            print(f"❌ Eleven Labs API error for scene {s_id}: {e}", flush=True)
            print(f"    Response: {response.text if 'response' in locals() else 'N/A'}", flush=True)
        except subprocess.CalledProcessError as e:
            print(f"❌ FFmpeg error trimming/padding for scene {s_id}: {e}", flush=True)
            print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        except Exception as e:
            print(f"❌ Unexpected error for scene {s_id}: {e}", flush=True)

    if not individual_audio_files:
        print("⚠️ No audio files were successfully generated. Aborting merge.", flush=True)
        return

    # 3. Concatenate all processed audio files
    combined_audio_path = os.path.join(target_folder_path, "_combined_tts_audio.mp3")
    concat_list_file = os.path.join(temp_audio_dir, "concat_tts_list.txt")

    with open(concat_list_file, "w") as f:
        for audio_f in individual_audio_files: f.write(f"file '{os.path.basename(audio_f)}'\n")

    print(f"\n🎶 Concatenating {len(individual_audio_files)} audio tracks...", flush=True)
    ffmpeg_concat_cmd = [
        "ffmpeg", "-y", "-f", "concat", "-safe", "0",
        "-i", concat_list_file,
        "-c", "copy", combined_audio_path
    ]
    try:
        subprocess.run(ffmpeg_concat_cmd, capture_output=True, text=True, check=True, cwd=temp_audio_dir)
        print(f"✅ All TTS audio tracks combined into: {combined_audio_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg error concatenating audio: {e}", flush=True)
        print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
        return

    # 4. Merge combined audio with master video
    print(f"\n🎬 Merging combined audio with master video: {master_video_path}", flush=True)
    ffmpeg_merge_cmd = [
        "ffmpeg", "-y",
        "-i", master_video_path,
        "-i", combined_audio_path,
        "-map", "0:v:0",  # Map video stream from master video
        "-map", "1:a:0",  # Map audio stream from combined audio
        "-c:v", "copy",   # Copy video stream directly
        "-c:a", "aac",    # Re-encode audio to AAC
        "-shortest",      # Ensure output duration is shortest of input streams
        output_video_with_tts_path
    ]

    try:
        subprocess.run(ffmpeg_merge_cmd, capture_output=True, text=True, check=True)
        print(f"\n🏆 MASTER VIDEO WITH TTS AUDIO SAVED TO: {output_video_with_tts_path}", flush=True)
    except subprocess.CalledProcessError as e:
        print(f"❌ FFmpeg error merging video and audio: {e}", flush=True)
        print(f"    FFmpeg stdout: {e.stdout}\n    FFmpeg stderr: {e.stderr}", flush=True)
    except Exception as e:
        print(f"❌ An unexpected error occurred during final merge: {e}", flush=True)
    finally:
        # Clean up temporary audio directory
        if os.path.exists(temp_audio_dir):
            try:
                shutil.rmtree(temp_audio_dir)
                print(f"✅ Cleaned up temporary audio directory: {temp_audio_dir}", flush=True)
            except Exception as e:
                print(f"⚠️ Error cleaning up temp audio directory {temp_audio_dir}: {e}", flush=True)


# --- Example Usage for the specified folder ---
# Define the base project root (adjust if your structure changes)
PROJECT_ROOT_CONFIG = "Colab Notebooks/MEDIAUPSCALE_FACTORY/Endless_Summers_Paradise"
BASE_PATH = f"/content/drive/MyDrive/{PROJECT_ROOT_CONFIG}"

# The specific project folder you want to process
TARGET_FOLDER = os.path.join(BASE_PATH, "CinemaScope_Syrup_Vortex_20260416_233406")

print(f"\nAttempting to generate and merge TTS audio for: {TARGET_FOLDER}\n", flush=True)
generate_and_merge_tts_for_folder(TARGET_FOLDER)
````

Now that you have the `generate_and_merge_elevenlabs_audio` function, you can integrate it into your `run_factory` function. Here's how you might modify the `run_factory` to use it:

1.  **Add `ELEVENLABS_VOICE_ID` to `userdata.get()`** in the `run_factory` function. I've used a default voice ID (`21m00Tcm4obs8fb3rGRf`) in the function for now, but you should set your preferred one in secrets.
2.  **Call the new function** after a scene's video has been successfully generated and saved to Drive.

Here's an example of how you'd modify the video saving part of `run_factory`:

```python
# Inside the PRODUCTION LOOP, after the video is saved to `v_file`:
if os.path.exists(v_file) and os.path.getsize(v_file) > 1000:
    # Check if audio is enabled in PROJECT_CONFIG
    if cfg['audio_enabled']:
        final_video_with_audio_path = f"{path}/scene_{s_id}_VIDEO_WITH_AUDIO.mp4"
        audio_prompt_text = scene.get('audio_prompt', '') # Get audio prompt from scene
        
        # Pass Eleven Labs API key (and optionally voice ID) if not None
        elevenlabs_voice_id = userdata.get('ELEVENLABS_VOICE_ID', '21m00Tcm4obs8fb3rGRf') # Default if not found

        if audio_prompt_text:
            merged_video_path = generate_and_merge_elevenlabs_audio(
                v_file,
                audio_prompt_text,
                final_video_with_audio_path,
                elevenlabs_voice_id,
                elevenlabs_api_key=userdata.get('ELEVENLABS_API_KEY')
            )
            if merged_video_path:
                video_paths.append(merged_video_path) # Add the video with audio to the list
                print(f"    ✅ Video with audio for scene {s_id} saved to Drive.", flush=True)
            else:
                video_paths.append(v_file) # If merge failed, still use the video without audio
                print(f"    ⚠️ Audio merge failed for scene {s_id}. Using video without audio.", flush=True)
        else:
            video_paths.append(v_file) # No audio prompt, just add the video
            print(f"    ⚠️ No audio prompt for scene {s_id}. Using video without audio.", flush=True)
    else:
        video_paths.append(v_file) # Audio disabled, just add the video
        print(f"    ✅ Video for scene {s_id} saved to Drive (audio disabled).")



```

This modification ensures that if `audio_enabled` is `True` and an `audio_prompt` exists, the Eleven Labs audio will be generated and merged. Remember to add `ELEVENLABS_API_KEY` and `ELEVENLABS_VOICE_ID` to your Colab secrets.
